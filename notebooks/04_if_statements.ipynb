# Estructuras condicionales

Las estructuras condicionales permiten **ejecutar instrucciones distintas** según las características de un dato.

Por ejemplo, en un flujo de trabajo SIG, se pueden utilizar en:

- identificación de cotas elevadas
- comprobación de coordenadas
- clasificar coberturas del suelo
- detectar registros incompletos.

## Primera condición: `if`

La instrucción `if` comprueba una **condición**. Si su resultado es `True` (verdadero), se ejecuta el bloque de código situado debajo. Si es `False` (falso), ese bloque se omite.

La sintaxis requiere **dos puntos (`:`)** al final de la condición y **sangría** en las líneas del bloque. En Python se utilizan habitualmente cuatro espacios por nivel de sangría (recomendación dentro de la PEP 8).

````python
elevation_m = 1450

if elevation_m > 1000:
    print("High-elevation location")

print("Elevation check completed")
````

En este ejemplo, `1450 > 1000` es `True` y se imprime el primer mensaje. La última instrucción se ejecuta siempre porque está fuera del bloque `if`.

Si la condición es falsa, Python no muestra ningún error: simplemente continúa con la siguiente instrucción que queda fuera del bloque.

````python
elevation_m = 650

if elevation_m > 1000:
    print("High-elevation location")

print("Elevation check completed")
````

## Condiciones y valores booleanos

Una comparación produce un valor de tipo `bool`, que puede ser `True` o `False` (con mayúscula inicial). Puede almacenarse en una variable y utilizarse después en un `if`.

````python
elevation_m = 1200
is_high_elevation = elevation_m > 1000

print(is_high_elevation)
print(type(is_high_elevation))

if is_high_elevation:
    print("Location exceeds the elevation threshold")
````

### Operadores de comparación

| Operador | Significado | Ejemplo |
|:--:|---|---|
| `==` | Igual a | `land_cover == "forest"` |
| `!=` | Distinto de | `crs_code != "EPSG:4326"` |
| `>` | Mayor que | `slope_deg > 30` |
| `>=` | Mayor o igual que | `elevation_m >= 1000` |
| `<` | Menor que | `rainfall_mm < 300` |
| `<=` | Menor o igual que | `distance_km <= 5` |

**Importante:** `=` asigna un valor; `==` compara dos valores. Una condición como `if elevation_m = 1000:` es incorrecta.

Se recomienda dejar un espacio a ambos lados de los operadores.

````python
elevation_m = 1000
land_cover = "forest"

print(elevation_m == 1000)
print(elevation_m != 800)
print(elevation_m > 1000)
print(elevation_m >= 1000)
print(elevation_m < 1500)
print(elevation_m <= 1000)
print(land_cover == "forest")
````

### Comparar textos

Las cadenas de texto se comparan teniendo en cuenta mayúsculas y minúsculas. Cuando los datos proceden de distintas fuentes, puede ser conveniente normalizar el texto antes de compararlo con `.lower()`. Recordar que este método devuelve una versión en minúsculas, sin modificar la variable original.

````python
land_cover = "Forest"

print(land_cover == "forest")
print(land_cover.lower() == "forest")
````

### Tipos de datos

Un número y una cadena que contienen los mismos caracteres **no son equivalentes**. Esto puede ocurrir al importar códigos o medidas como texto. La conversión con `int()` o `float()` solo es posible si el contenido admite una interpretación numérica.

````python
elevation_text = "1200"
elevation_m = 1200

print(elevation_text == elevation_m)
print(int(elevation_text) == elevation_m)

# Keep identifiers as strings when their leading zeros matter.
zone_code = "005"
print(zone_code == "005")
````

## Pertenencia: `in` y `not in`

Los operadores `in` y `not in` permiten comprobar si un elemento pertenece a una lista, tupla o conjunto. Resultan útiles para validar categorías o códigos presentes en un conjunto de datos.

````python
allowed_land_covers = {"forest", "cropland", "water", "urban"}
land_cover = "wetland"

print(land_cover in allowed_land_covers)
print(land_cover not in allowed_land_covers)

if land_cover not in allowed_land_covers:
    print("Unknown land-cover category")
````

## Elegir entre dos alternativas: `if`-`else`

La cláusula `else` define qué hacer cuando la condición del `if` es falsa. **Solo se ejecuta uno de los dos bloques.** `else` no lleva condición propia.

````python
distance_km = 3.2

if distance_km <= 5:
    print("Site is within the study buffer")
else:
    print("Site is outside the study buffer")
````

El ejemplo usa una distancia calculada previamente. La comprobación determina si un emplazamiento queda dentro de un umbral de 5 km.

## Varias alternativas excluyentes: `if`-`elif`-`else`

Cuando existen más de dos resultados posibles, se utiliza `elif` (abreviatura de *else if*). Python evalúa las condiciones **en orden** y ejecuta **el bloque de la primera condición verdadera**. A partir de ese momento no examina las demás condiciones de la misma cadena.

La cláusula `else` es opcional. Si existe, se ejecuta cuando ninguna condición anterior es verdadera. Puede haber varios `elif`, pero como máximo un `else`.

````python
slope_deg = 18

if slope_deg >= 30:
    slope_class = "steep"
elif slope_deg >= 15:
    slope_class = "moderate"

print(slope_class)
````

**El orden importa.** Si primero se comprueba `slope_deg >= 15`, un valor de 35 grados se clasificaría como pendiente moderada y nunca llegaría a la condición de pendiente fuerte.

````python
slope_deg = 35

if slope_deg >= 15:
    slope_class = "moderate"
elif slope_deg >= 30:
    slope_class = "steep"

print(slope_class)
````

Los ejemplos anteriores crean una variable en función de la condición. Hay que tener en cuenta que la variable solo se creará cuando la condición se cumpla. Al no tener una cláusula `else`, si el valor de la pendiente no se corresponde con las dos condiciones, la variable nunca será creada y la última línea dará un error `NameError`.

Por lo tanto, es muy importante utilizar una cláusula `else` que permita guardar la variable.

**Error intencionado: `NameError`.** En la siguiente celda, la pendiente no cumple ninguna de las dos condiciones.

````python
slope_deg = 5

if slope_deg >= 30:
    slope_cls = "steep"
elif slope_deg >= 15:
    slope_cls = "moderate"

print(slope_cls)
````

````python
slope_deg = 5

if slope_deg >= 30:
    slope_cls = "steep"
elif slope_deg >= 15:
    slope_cls = "moderate"
else:
    slope_cls = "low"

print(slope_cls)
````

### Tratar explícitamente valores ausentes

En un conjunto de datos puede aparecer `None` para indicar que falta un valor. Conviene comprobarlo **antes** de realizar comparaciones numéricas; comparar `None` con un número mediante `>` o `<` provoca un error de tipo (`TypeError`).

**Error intencionado: `TypeError`.** La siguiente celda compara `None` con un número.

````python
elevation_m = None

if elevation_m >= 1000:
    print("High-elevation location")
else:
    print("Lower-elevation location")
````

````python
elevation_m = None

if elevation_m is None:
    print("Missing elevation")
elif elevation_m >= 1000:
    print("High-elevation location")
else:
    print("Lower-elevation location")
````

Para comprobar `None` se utilizan `is None` e `is not None`.

**Importante**: El número `0` es un valor numérico válido y no debe confundirse con la ausencia de datos.

El operador `is` comprueba la *identidad*: detecta si dos nombres se refieren al mismo objeto en memoria, no si son iguales o del mismo tipo:

- `None is None` es `True` porque Python solo tiene un objeto `None`, así que cada `None` en Python es el mismo objeto. 
- `14 is int` es `False` porque `14` es un objeto de tipo `int`, mientras que `int` es la clase en sí misma, por lo que son dos objetos distintos.

Si se quiere comprobar el tipo de un objeto, se utiliza la función `isinstance()`. Como norma, utiliza `is` solamente para comprobar si un objeto es `None` y usa `==` para el resto de comparaciones.

> ¿Qué significa que `None` es un objeto único en Python? Significa que solo hay un objeto `None` que Python crea al iniciarse, y cada vez que se escribe `None` se crea una referencia a ese primer objeto. Python nunca crea un segundo `None`.

````python
slope_mean = 100000
slope_max = 100000

print("is", slope_mean is slope_max)
print("==", slope_mean == slope_max)
print("is int", slope_max is int)
print("isinstance", isinstance(slope_max, int))
````

## Combinar condiciones: `and`, `or` y `not`

Los operadores lógicos permiten construir condiciones más precisas:

- `and`: verdadero únicamente cuando **todas** las condiciones son verdaderas.
- `or`: verdadero cuando **al menos una** condición es verdadera.
- `not`: invierte el valor booleano de una condición (por ejemplo, `not True` devuelve `False`).


````python
elevation_m = 1250
slope_deg = 22

if elevation_m >= 1000 and slope_deg >= 20:
    print("High and steep location")

land_cover = "water"
if land_cover == "water" or land_cover == "wetland":
    print("Aquatic or wetland category")

has_coordinates = False
if not has_coordinates:
    print("Coordinates need to be checked")
````

### Intervalos y prioridades

Python admite comparaciones encadenadas, como `500 <= elevation_m < 1500`, equivalentes a comprobar ambos límites con `and`.

````python
elevation_m = 800

is_moderate = 500 <= elevation_m < 1500
is_moderate_v2 = elevation_m >= 500 and elevation_m < 1500

print(is_moderate)
print(is_moderate_v2)
````

En expresiones lógicas mixtas, Python evalúa los operadores en un orden fijo de precedencia:

1. primero `not`,
2. luego `and`,
3. y por último `or`.

`and` "une" sus *operandos* antes que `or`, de modo que la siguiente expresión

```text
a or b and c
```

se interpreta como `a or (b and c)`, y no como `(a or b) and c`.

Como ambas lecturas pueden dar resultados distintos, es fácil que el código se comporte de forma diferente a lo previsto. Por ejemplo, el siguiente bloque de código debe mostrar un mensaje **solo** si existe un área protegida, y si esta tiene un tipo de cobertura `forest` o `wetland`. 

````python
land_cover = "forest"
protected_area = False

if land_cover == "forest" or land_cover == "wetland" and protected_area:
    print("Priority area for field review")
````

A pesar de que el no es un área protegida (`protected_area = False`) el bloque `if` ha detectado que se cumple la condición y mostrado el mensaje.

¿Por qué? Porque la parte con `and` se evalúa por separado. Python evalúa la condición anterior de esta forma:

```python
(land_cover == "forest") or ((land_cover == "wetland") and protected_area)
```

Para evitar estas ambigüedades, se recomienda usar paréntesis siempre que se combinen `and` y `or`: aunque no siempre cambien el resultado, dejan clara la intención y hacen el código más fácil de leer y de mantener.


````python
land_cover = "forest"
protected_area = False

if (land_cover == "forest" or land_cover == "wetland") and protected_area:
    print("Priority area for field review")
````

### Evaluación de cortocircuito

Python evalúa las expresiones `and` y `or` de izquierda a derecha y puede detenerse cuando el resultado ya está decidido:

- Con `and`, si la primera condición es falsa, no necesita evaluar la segunda.
- Con `or`, si la primera condición es verdadera, no necesita evaluar la segunda.

Esta propiedad permite comprobar un dato antes de compararlo con un número. En el siguiente ejemplo, si no existiera esta funcionalidad, la condición devolvería un error debido a que `None` no puede compararse con un número.

````python
elevation_m = None

if elevation_m is not None and elevation_m > 1000:
    print("High-elevation location")
else:
    print("Elevation is missing or does not exceed the threshold")
````

**Error intencionado: `TypeError`.** Si se invierte el orden de las condiciones, Python evalúa primero `elevation_m > 1000` y compara `None` con un número.

````python
elevation_m = None

if elevation_m > 1000 and elevation_m is not None:
    print("High-elevation location")
else:
    print("Elevation is missing or does not exceed the threshold")
````

## Varias comprobaciones independientes

Una cadena `if`-`elif` solo ejecuta un bloque, el que cumple la condición. En cambio, varios `if` independientes permiten que se ejecuten **varios bloques** cuando distintas condiciones son verdaderas. Es útil para señalar diferentes problemas, por ejemplo al evaluar una capa antes de ser utilizada.

````python
latitude = 95.0
longitude = -0.89
crs_code = "EPSG:4326"

if not -90 <= latitude <= 90:
    print("Latitude outside valid range")

if not -180 <= longitude <= 180:
    print("Longitude outside valid range")

if crs_code != "EPSG:4326":
    print("Unexpected coordinate reference system")
````

La siguiente versión utiliza `elif` para las mismas comprobaciones. Solo podría mostrar el **primer** problema encontrado, incluso aunque hubiera varios.

````python
latitude = 95.0
longitude = 190.0

if not -90 <= latitude <= 90:
    print("Latitude outside valid range")
elif not -180 <= longitude <= 180:
    print("Longitude outside valid range")
````

## Valores que se interpretan como verdaderos o falsos

Las condiciones no siempre contienen una comparación explícita. Python considera falsos los siguientes valores:

- `False`
- `None`
- el número `0`
- la cadena vacía `""`
- las colecciones vacías como `[]`, `()` o `set()`

Los números distintos de cero y las colecciones con elementos suelen interpretarse como verdaderos.

Esto resulta práctico para comprobar si hay elementos, pero **no** para distinguir un valor cero de un dato ausente.

````python
selected_features = []

if selected_features:
    print("There are selected features")
else:
    print("No features selected")

if len(selected_features) > 0:
    print("There are selected features")
else:
    print("No features selected")
````

````python
selected_features = [101, 102, 108]
if selected_features:
    print("There are selected features")
````

````python
elevation_m = 0

if elevation_m:
    print("Elevation has a non-zero value")
else:
    print("Elevation is zero or the value is falsy")

# An explicit check distinguishes zero from missing data.
if elevation_m is not None:
    print("Elevation is available")
````

## Errores frecuentes

| Error | Consecuencia o corrección |
|---|---|
| Usar `=` en lugar de `==` | `=` asigna; `==` compara. |
| Omitir `:` al final de `if` o `elif` | Produce un error de sintaxis. |
| No aplicar sangría al bloque | Produce un error de sangría o cambia el flujo del programa. |
| Comprobar umbrales solapados en orden incorrecto | La primera condición verdadera impide comprobar las siguientes de una cadena `elif`. |
| Utilizar varios `if` cuando se necesita una sola categoría | Pueden ejecutarse varias ramas. |
| Comparar `None` con un número | Puede producir un `TypeError`; primero se comprueba si falta el dato. |
| Utilizar `if elevation_m:` para detectar un dato válido | Confunde la elevación `0` con un dato ausente. |
| Comparar un número con texto numérico | Primero debe convertirse el texto, si procede. |

## Síntesis

- `if` ejecuta un bloque cuando se cumple una condición.
- `else` define la alternativa; `elif` añade condiciones que se comprueban por orden.
- Varios `if` independientes sirven para comprobaciones que pueden cumplirse simultáneamente.
- `==`, `!=`, `>`, `>=`, `<`, `<=`, `in` y `not in` permiten construir condiciones.
- `and`, `or` y `not` permiten combinarlas.
- La distinción entre `0`, `None` y colecciones vacías es importante al tratar datos geográficos.
