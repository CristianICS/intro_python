# Introducción a las funciones

Uno de los principios básicos de la programación es "No te repitas" (*Don't Repeat Yourself*). Si una acción debe ejecutarse muchas veces, se define una sola vez y se invoca cada vez que sea necesaria.

En un flujo de trabajo relacionado con información espacial esto es habitual: aplicar el mismo procesamiento a muchas capas, validar muchas coordenadas o convertir muchas unidades. Las funciones reducen el trabajo y **los errores**.

## ¿Qué son las funciones?

Una función es un conjunto de instrucciones agrupadas bajo un nombre. Ya se han usado funciones y métodos del lenguaje, como `print()`, `string.title()` o `list.sort()`. También se pueden definir funciones propias.

## Sintaxis general

Una función genérica tiene este aspecto (es un esquema, no código ejecutable):

```python
# Definición de la función
def function_name(parameter_1, parameter_2):
    # Instrucciones que usan parameter_1 y parameter_2

# Llamada a la función
function_name(value_1, value_2)
```

### Definir una función

- Se escribe la palabra clave `def`, que indica a Python que se va a *definir* una función.
- Se elige un nombre. El nombre de una variable indica qué contiene; el de una función debe indicar qué *hace* (por ejemplo, `calculate_area`).
- Entre paréntesis se escriben los *parámetros*: nombres de variable que reciben los valores que la función necesita. **Solo existen dentro de la función** y pueden diferir de los nombres usados en el resto del programa.
- La línea de definición termina con dos puntos (`:`).
- El cuerpo de la función se escribe con sangría debajo de esa línea.

### Usar (llamar a) una función

- Se escribe su nombre seguido de paréntesis.
- Dentro de los paréntesis se indican los *argumentos*: los valores concretos con los que trabajará la función. Pueden ser variables o valores literales.
- Resumen: los **parámetros** aparecen en la definición; los **argumentos** aparecen en la llamada.

## Ejemplos básicos

En el primer ejemplo, un programa informa del procesamiento de varias capas. Se realiza de forma manual, sin funciones.

````python
print("Processing layer: Roads")
print("Checking that all geometries are valid.")

print("\nProcessing layer: Rivers")
print("Checking that all geometries are valid.")

print("\nProcessing layer: Municipalities")
print("Checking that all geometries are valid.")
````

Las dos instrucciones `print` se repiten tres veces y solo cambia el nombre de la capa.

Se puede concentrar el código repetido en un único lugar con una función:

````python
def process_layer(layer_name):
    """Print a two-line status message for a layer."""
    print(f"\nProcessing layer: {layer_name}")
    print("Checking that all geometries are valid.")


process_layer("Roads")
process_layer("Rivers")
process_layer("Municipalities")
````

La función se llama `process_layer()` y tiene un parámetro, `layer_name`, que recibe el valor indicado en cada llamada. Aquí se llama tres veces con argumentos distintos.

El comentario inicial de la función, escrito con tres comillas (`"""`) se llama *docstring*. Es un tipo de comentario especial que puede consultarse como una ayuda. La función `help()` permite consultar esta información incluyendo el nombre de la función como argumento.

````python
help(process_layer)
````

### Estilo de las funciones

Las funciones siguen dos guías de estilo:

- La PEP 8, vista en el cuaderno `01_estilo_comentarios.ipynb`, recomienda dejar **dos líneas en blanco** antes y después de la definición de una función. Así, la función queda separada con claridad del resto del código.
- La [PEP 257](https://peps.python.org/pep-0257/) recoge las convenciones de los *docstrings*. Recomienda escribirlos en modo imperativo, como una orden: `"""Print a message."""` en lugar de `"""Prints a message."""`, y terminarlos con un punto.

Los ejemplos de este cuaderno siguen ambas recomendaciones.

## Un error común

Una función debe estar definida **antes de ser llamada**. Si la definición queda al final del programa, Python lanza un `NameError`:

````python
process_layer_late("Roads")


def process_layer_late(layer_name):
    print(f"\nProcessing layer: {layer_name}")
    print("Checking that all geometries are valid.")
````

Python lee el código de arriba abajo; en la primera línea aún no conoce la función. Por ello, las funciones se definen al principio del programa.

## Un segundo ejemplo

Imprimir una lista con un bucle `for` requiere dos líneas que se repiten cada vez que se quiere mostrar su contenido. Primero, el código sin función:

````python
rivers = ["guadalquivir", "ebro", "tajo"]

# Put rivers in alphabetical order.
rivers.sort()
print("Rivers in alphabetical order.")
for river in rivers:
    print(river.title())

# Put rivers in reverse alphabetical order.
rivers.sort(reverse=True)
print("\nRivers in reverse alphabetical order.")
for river in rivers:
    print(river.title())
````

Y ahora con una función que imprime la lista:

````python
def show_rivers(rivers, message):
    """Print a message followed by the list of rivers."""
    print(f"\n{message}")
    for river in rivers:
        print(river.title())


rivers = ["guadalquivir", "ebro", "tajo"]

rivers.sort()
show_rivers(rivers, "Rivers in alphabetical order.")

rivers.sort(reverse=True)
show_rivers(rivers, "Rivers in reverse alphabetical order.")
````

El código es más limpio. La función necesita dos datos (la lista y el mensaje) y el bucle dentro de ella es idéntico al de la versión anterior. El resto del programa se centra en lo que cambia (ordenar la lista) y no en cómo se imprime.

## Ventajas de usar funciones

- **Se escribe una vez.** Se ahorra trabajo, y más en programas grandes.
- **Menos errores.** Cada repetición de código es una oportunidad de equivocarse. Con una función hay un solo lugar donde corregir y, una vez corregido, el comportamiento es fiable en todas las llamadas.
- **Los cambios se propagan.** Modificar la función cambia el comportamiento en todas las llamadas, sin editar el programa en muchos sitios.

Por ejemplo, para mostrar la lista con guiones basta cambiar una línea dentro de la función:

````python
def show_rivers(rivers, message):
    """Print a message followed by the list of rivers, as a bulleted list."""
    print(f"\n{message}")
    for river in rivers:
        print(f"- {river.title()}")


rivers = ["guadalquivir", "ebro", "tajo"]

rivers.sort()
show_rivers(rivers, "Rivers in alphabetical order.")

rivers.sort(reverse=True)
show_rivers(rivers, "Rivers in reverse alphabetical order.")
````

## Devolver un valor

Una función puede **devolver** un valor con `return`.

La siguiente función recibe una altitud en metros y devuelve una categoría de relieve:

````python
def classify_elevation(elevation):
    """Take an elevation in meters and return a relief category."""
    if elevation < 0:
        return "below sea level"
    elif elevation <= 200:
        return "lowland"
    elif elevation <= 1000:
        return "hills"


for current_elevation in [-5, 150, 700, 2500]:
    category = classify_elevation(current_elevation)
    print(current_elevation, category)
````

Este programa no lanza ningún error de sintaxis, pero tiene un **error lógico**: para 2500 m el resultado es `None`. Cuando una función termina sin ejecutar ningún `return`, devuelve `None` (la ausencia de valor).

Recordar que el método `append()`, utilizado para añadir elementos a una lista, devolvía `None` por el mismo motivo.

La solución es añadir un caso final con `else` que cubra el resto de valores:

````python
def classify_elevation(elevation):
    """Take an elevation in meters and return a relief category."""
    if elevation < 0:
        return "below sea level"
    elif elevation <= 200:
        return "lowland"
    elif elevation <= 1000:
        return "hills"
    else:
        return "mountain"


for current_elevation in [-5, 150, 700, 2500]:
    category = classify_elevation(current_elevation)
    print(current_elevation, category)
````

La función **termina en cuanto ejecuta un `return`**. Cualquier línea posterior no se ejecuta nunca:

````python
def classify_elevation(elevation):
    if elevation < 0:
        return "below sea level"
    else:
        return "above sea level"

    # This line never executes: the function has already returned a value.
    print("This message will never be printed.")


print(classify_elevation(100))
````

### `print()` frente a `return`

Es una confusión muy frecuente:

- `print()` **muestra** un valor en pantalla, pero el programa no puede reutilizarlo.
- `return` **entrega** un valor al código que llamó a la función, que puede guardarlo en una variable, operar con él o imprimirlo.

Un ejemplo útil en cartografía es obtener la zona UTM (husos de 6° de longitud, numerados desde −180°) en la que se encuentra un punto. La siguiente función es una versión simplificada que ignora las excepciones de Noruega y Svalbard.

````python
def get_utm_zone(longitude):
    """Take a longitude in decimal degrees and return the UTM zone number."""
    return int((longitude + 180) / 6) + 1


madrid_zone = get_utm_zone(-3.7)
barcelona_zone = get_utm_zone(2.17)

print("Madrid is in UTM zone", madrid_zone)
print("Barcelona is in UTM zone", barcelona_zone)
print("Zone difference:", barcelona_zone - madrid_zone)
````

## Variables locales

Los parámetros y las variables creadas dentro de una función son **locales**: solo existen mientras la función se ejecuta. Por eso hay que usar `return` para sacar resultados. Esta celda falla a propósito:

````python
def calculate_area(width, height):
    area = width * height


calculate_area(3, 4)
print(area)  # NameError: 'area' only exists inside the function
````

## Resumen

- Una función agrupa instrucciones bajo un nombre. Se define con `def` y se llama escribiendo su nombre seguido de paréntesis.
- Los **parámetros** aparecen en la definición; los **argumentos**, en la llamada.
- Una función debe definirse antes de ser llamada.
- El *docstring* describe qué hace la función y puede consultarse con `help()`.
- `return` entrega un valor al código que llama a la función. Si no se ejecuta ningún `return`, la función devuelve `None`.
- `print()` solo muestra un valor; `return` permite reutilizarlo.
- Las variables creadas dentro de una función son **locales**: no existen fuera de ella.
