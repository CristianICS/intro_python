# Variables y tipos de datos

En este capítulo se aprende a guardar información, reconocer sus tipos y realizar operaciones sencillas con Python. No se requieren conocimientos previos de programación. Como referencia, se muestran algunas similitudes con los ejemplos básicos de R.

**Objetivos de aprendizaje**

- Guardar y consultar valores mediante variables.
- Distinguir números enteros, números decimales, textos y valores lógicos.
- Escribir nombres de variables claros.
- Realizar operaciones con números y textos.
- Convertir datos y presentar resultados de forma legible.

## 1. Primeras instrucciones

Un **programa** es un conjunto de instrucciones que sigue el ordenador. En Python, una instrucción puede mostrar un mensaje.

`print()` es una **función**: una instrucción reutilizable que realiza una tarea. En este caso, muestra el valor indicado entre paréntesis. El texto se escribe entre comillas.

````python
print("Hola, mundo")
````

## 2. Guardar información en variables

Una **variable** es un nombre que permite guardar un valor y recuperarlo después. El signo `=` **asigna** el valor de la derecha al nombre de la izquierda.

```python
nombre = valor
```

No es una igualdad matemática: se utiliza para guardar información. En los ejemplos siguientes, `message` es el nombre y `"Hola, mundo"` es el valor.

````python
message = "Hola, mundo"
print(message)
````

````python
message = "Primer mensaje"
print(message)

message = "Segundo mensaje"
print(message)
````

La siguiente asignación en R

```r
n <- 4
```

se escribe de la siguiente forma en Python:

```python
n = 4
```

Ambos ejemplos guardan un valor con un nombre.

## 3. Tipos de datos

Un **tipo de dato** indica qué clase de información representa un valor y qué operaciones admite. Se utilizarán cuatro tipos básicos:

| Tipo en Python | Significado | Ejemplo |
|---|---|---|
| `int` | Número entero | `25` |
| `float` | Número decimal | `3.5` |
| `str` | Texto o cadena de caracteres | `"Zaragoza"` |
| `bool` | Valor lógico | `True` o `False` |

Una **cadena de caracteres** es una sucesión de letras, números, espacios u otros símbolos tratada como texto. Un **valor lógico** expresa si una afirmación es verdadera (`True`) o falsa (`False`). Estas palabras se escriben con inicial mayúscula.

La función `type()` permite consultar el tipo de un valor. Python, gracias a su tipado dinámico, identifica el tipo al asignarlo; no es necesario declararlo previamente.

````python
age = 25
temperature = 18.5
city = "Zaragoza"
is_capital = True

print(type(age))
print(type(temperature))
print(type(city))
print(type(is_capital))
````

Una misma variable puede recibir valores de tipos distintos en momentos diferentes. Es preferible mantener un significado coherente para facilitar la lectura.

````python
value = "15"
print(type(value))

value = 15
print(type(value))
````

## 4. Nombres de variables

Un nombre válido puede:

- contener letras, números y guiones bajos (`_`)
- **no puede** comenzar por un número.
- **no puede** contener espacios ni guiones medios (`-`)
- **no puede** coincidir con una **palabra reservada** de Python, como `for` o `class`.

La guía de estilo [PEP 8](https://peps.python.org/pep-0008/) recomienda:

- Utilizar nombres descriptivos, como `student_age`.
- Escribir los nombres en minúsculas y separar las palabras con `_`. Este estilo se denomina *snake_case*.
- Dejar un espacio a cada lado de `=`.
- Evitar nombres que puedan confundirse, como una `l` minúscula aislada (que puede confundirse con un 1) o una `O` mayúscula aislada (con un 0).

Se utilizan nombres en inglés en los ejemplos por coherencia con el código habitual; no es una obligación del lenguaje.

### Ejercicio 1. Mejorar nombres

Las siguientes líneas muestran nombres poco claros o incorrectos. **No es necesario ejecutarlas.** Se propone reescribirlas con nombres descriptivos y válidos.

```python
a = 25
UserName = "Alice"
total-price = 149.99
string = "chair"
```

## 5. Comprender un error: `NameError`

Un **error** aparece cuando Python no puede ejecutar una instrucción. El mensaje indica qué ha sucedido y en qué línea. `NameError` suele indicar que se está utilizando un nombre que Python no conoce.

### Ejercicio 2. Error intencionado

En la siguiente celda, el nombre `mesage` contiene una errata. Se recomienda ejecutarla, leer el mensaje de error y después continuar con la solución.

````python
message = "Un mensaje nuevo"
print(mesage)
````

El mensaje `NameError` señala que `mesage` no está definido. El nombre asignado había sido `message`, con dos letras `s`. Python considera que son nombres diferentes.

**Solución:** escribir el mismo nombre en la asignación y en la llamada a `print()`.

````python
message = "Un mensaje nuevo"
print(message)
````

## 6. Trabajar con texto (`str`)

El texto se escribe entre comillas simples (`'`) o dobles (`"`). Ambas formas permiten crear el mismo tipo de dato. Se recomienda elegir una forma y utilizarla de manera uniforme.

````python
city = "Zaragoza"
country = 'Spain'

print(city)
print(country)
````

Es posible incluir comillas simples dentro de un texto delimitado por comillas dobles, o al contrario.

````python
sentence = "El cartel dice 'Bienvenidos'."
print(sentence)
````

### Métodos de texto

Un **método** es una operación asociada a un tipo de dato. Para utilizarlo, se escribe un punto después del valor o de la variable y, a continuación, el nombre del método con paréntesis.

Por ejemplo, `.upper()` crea una versión del texto en mayúsculas; `.lower()` la crea en minúsculas y `.title()` aplica mayúsculas iniciales a las palabras.

````python
first_name = "ada"

print(first_name.upper())
print(first_name.lower())
print(first_name.title())
print(first_name)
````

Estos métodos devuelven un texto nuevo: **no cambian** la cadena guardada en `first_name`. Para conservar el resultado, se puede asignar a otra variable.

````python
first_name = "ada"
formatted_name = first_name.title()
print(formatted_name)
````

### Unir cadenas de texto

**Concatenar** significa unir cadenas de texto. En Python se utiliza `+`. Los espacios que deben aparecer en el resultado han de incluirse expresamente.

````python
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name

print(full_name)
````

### Insertar valores en un texto

Una **cadena formateada** o *f-string* permite insertar el valor de una variable dentro de un texto. Se coloca `f` antes de las comillas y el nombre de la variable entre llaves `{}`.

Este procedimiento facilita la presentación de resultados sin tener que unir fragmentos con `+`.

````python
city = "Zaragoza"
population = 682513

print(f"{city} tiene {population} habitantes.")
````

Otra posibilidad es utilizar el método `.format()`. Las llaves actúan como **marcadores de posición**: señalan dónde se colocará cada valor.

````python
template = "El municipio de {municipality} está en {province}."

print(template.format(municipality="Calatayud", province="Zaragoza"))
````

Si los marcadores están vacíos (`{}`), los valores se insertan en el orden indicado.

````python
template = "Coordenadas: {}, {}"
latitude = "37.24N"
longitude = "115.81W"

print(template.format(latitude, longitude))
````

### Espacios, tabulaciones y saltos de línea

Los **caracteres de espacio en blanco** incluyen espacios, tabulaciones y saltos de línea. Dentro de una cadena, `\t` representa una tabulación y `\n` un salto de línea.

````python
message = "Coordenadas:\n\tLatitud: 37.24N\n\tLongitud: 115.81W"
print(message)
````

En los datos procedentes de tablas pueden aparecer espacios sobrantes. Los métodos `.lstrip()`, `.rstrip()` y `.strip()` eliminan espacios en blanco del inicio, del final y de ambos extremos, respectivamente. No eliminan los espacios que estén en el centro.

````python
raw_text = "   37.24N  "

print(f"-{raw_text.lstrip()}-")
print(f"-{raw_text.rstrip()}-")
print(f"-{raw_text.strip()}-")
````

## 7. Trabajar con números

Python permite realizar operaciones aritméticas con los símbolos siguientes:

| Operación | Símbolo | Ejemplo |
|---|---|---|
| Suma | `+` | `4 + 2` |
| Resta | `-` | `4 - 2` |
| Multiplicación | `*` | `4 * 2` |
| División | `/` | `4 / 2` |
| Potencia | `**` | `4 ** 2` |

En Python, los números decimales utilizan un **punto** (`3.5`) en lugar de una coma (`3,5`).

````python
print(4 + 2)
print(4 - 2)
print(4 * 2)
print(4 / 2)
print(4 ** 2)
````

Como en matemáticas, la multiplicación tiene prioridad sobre la suma. Los paréntesis permiten cambiar el orden de las operaciones.

````python
standard_result = 2 + 3 * 4
parenthesized_result = (2 + 3) * 4

print(standard_result)
print(parenthesized_result)
````

### Números enteros (`int`) y decimales (`float`)

Un número sin parte decimal, como `1000`, se representa habitualmente con `int`. Un número con parte decimal, como `1.5`, se representa habitualmente con `float`.

El resultado de la división `/` es de tipo `float`, incluso cuando el cociente es un número entero.

````python
distance_m = 1000
distance_km = distance_m / 1000

print(type(distance_m))
print(distance_km)
print(type(distance_km))
````

### Precisión de los números decimales

Algunos números decimales no pueden representarse exactamente en el sistema binario que utiliza el ordenador. Por eso, una operación sencilla puede mostrar una pequeña diferencia respecto al resultado matemático.

````python
print(0.1 + 0.2)
print(0.1 + 0.1 + 0.1 - 0.3)
````

Es un efecto de **precisión numérica**, no un error en la fórmula. En cálculos con coordenadas u otros valores decimales, conviene tenerlo en cuenta cuando se necesite comparar resultados con mucha exactitud.

Lectura opcional: [información sobre números de punto flotante en Python](https://docs.python.org/3/tutorial/floatingpoint.html).

## 8. Convertir tipos de datos

**Convertir un tipo** significa obtener un valor equivalente representado de otra forma. Es habitual cuando un archivo contiene números guardados como texto. En Python se pueden convertir tipos de texto mediante las siguientes funciones:

- `int()` intenta obtener un número entero.
- `float()` intenta obtener un número decimal.
- `str()` obtiene una representación en texto.

El texto debe tener un formato numérico válido para convertirse a número.

````python
population_text = "682513"
population = int(population_text)

latitude_text = "41.6488"
latitude = float(latitude_text)

print(population, type(population))
print(latitude, type(latitude))
````

La función `int()` **elimina la parte decimal** de un número positivo, no lo redondea al entero más próximo. En general, trunca el valor hacia cero.

````python
elevation = 582.9
elevation_int = int(elevation)

print(elevation_int)
````

Para mostrar un número junto a un texto, una opción es convertirlo con `str()`. También puede utilizarse una cadena formateada (ver [Sección Trabajar con texto (str)](#insertar-valores-en-un-texto)).

````python
population = 682513
print("Población: " + str(population))
print(f"Población: {population}")
````

## 9. Comprender un error: `TypeError`

`TypeError` aparece cuando se intenta una operación que no admite el tipo de dato utilizado. Por ejemplo, `+` puede sumar números o unir cadenas, pero no puede unir directamente una cadena y un número entero.

> Igual que R, Python tiene un **tipado fuerte**: no permite realizar operaciones no válidas entre tipos incompatibles sin conversión explícita.

### Ejercicio 3. Error intencionado

Se recomienda ejecutar esta celda, observar el error y continuar con las dos soluciones.

````python
result = 2 + 2
print("Resultado: " + result)
````

El problema es que `"Resultado: "` es texto (`str`) y `result` es un número (`int`). Python no los concatena automáticamente. Se puede convertir el número con `str()` o insertarlo mediante una *f-string*.

````python
result = 2 + 2

print("Resultado: " + str(result))
print(f"Resultado: {result}")
````

## 10. Formato de números

Un **especificador de formato** controla cómo aparece un valor al mostrarlo, sin modificar el valor original. Se escribe después de `:` dentro de las llaves de una *f-string*.

Por ejemplo, `:.2f` muestra un número con dos cifras decimales. Otros formatos útiles son:

| Formato | Resultado que produce |
|---|---|
| `:.2f` | Dos cifras decimales |
| `:.0f` | Ninguna cifra decimal |
| `:,` | Separador de miles según la convención del formato |
| `:,.2f` | Separador de miles y dos decimales |
| `:.1%` | Porcentaje con una cifra decimal |

El formato numérico estándar de Python utiliza el punto decimal y, en `:,`, la coma como separador de miles; no adapta estos signos automáticamente a la convención española.

````python
pi_value = 3.141592653589793
print(f"Pi con dos decimales: {pi_value:.2f}")
print(f"Pi sin decimales: {pi_value:.0f}")
````

````python
surface = 1254321.567
print(f"Superficie: {surface:,.2f} metros cuadrados")

proportion = 0.253
print(f"Proporción: {proportion:.1%}")
````

El formato controla únicamente el texto mostrado. El valor almacenado conserva sus cifras originales.

````python
surface = 1254321.567
print(f"{surface:,.2f}")
print(surface)
````

## 11. Estilo de las operaciones

PEP 8 recomienda dejar un espacio alrededor de la mayoría de los operadores para facilitar la lectura. Cuando una expresión contiene operaciones con distinta prioridad, puede omitirse el espacio alrededor de las que tienen mayor prioridad si mejora la claridad.

````python
total = 10 + 5
area = 10 * 5
result = 10 + 5*2

print(total)
print(area)
print(result)
````
