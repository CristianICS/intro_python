# Estilo y comentarios

Antes de comenzar a escribir código de Python, es importante conocer las reglas de estilo que se seguirán durante todo el curso.

## Estilo de programación

Los desarrolladores de Python diseñaron en lenguaje teniendo en cuenta que el código **se lee con mucha más frecuencia de la que se escribe**. Un buen lenguaje tiene que ser fácil de leer y fácil de escribir. Parte del éxito de Python como lenguaje se debe a la legibilidad de su código. Las convenciones de escritura permiten a los futuros lectores del código, incluidas las personas encargadas de escribirlo, saber la finalidad para la que fue escrito.

> Una PEP contiene convenciones, no leyes ni sintaxis.

Una PEP es una *Python Enhancement Proposal* (propuesta de mejora de Python). Se redactan cuando quiere proponerse un cambio en el lenguaje, o incluir una recomendación. La [PEP 8](https://peps.python.org/pep-0008/) es una recopilación de pautas para escribir código fácil de leer. Por ahora, gran parte de su contenido puede resultar difícil de comprender. Los ejemplos y ejercicios de este curso siguen la PEP 8 y la PEP 257.

> Adquirir buenos hábitos de estilo desde el inicio ayuda a escribir código limpio y a entenderlo mejor.

### PEP 8

Contiene las recomendaciones generales para escribir código fácil de leer.

### Longitud de línea

- Longitud de línea de 79 caracteres máximo (72 para los comentarios)

Esta regla surgió en los primeros años de Python, cuando las pantallas y las terminales solo podían mostrar unos 80 caracteres por línea. Hoy las pantallas son mucho más grandes, pero la regla sigue siendo útil: las líneas cortas son más fáciles de leer, permiten comparar dos archivos uno al lado del otro y se ven bien en terminales como PowerShell, donde una línea demasiado larga se corta o se desplaza de forma incómoda.

```python
print("Imprimir el primer mensaje del curso utilizando una variable de tipo texto, demasiado larga como para incluirse en una sola linea")
```

La línea anterior contiene una longitud superior a 79 caracteres. La mejor opción sería dividirla en varias líneas.

```python
print(
    "Imprimir el primer mensaje del curso utilizando una variable de tipo "
    "texto, demasiado larga como para incluirse en una sola línea"
)
```

En Jupyter Lab (y en la mayoría de las IDEs) se puede [añadir una barra vertical](https://stackoverflow.com/q/49322559/23551600) que marque el límite de los 79 caracteres.

1. Abrir el editor de opciones (`Ctrl + ,`)
2. En el panel izquierdo, seleccionar *Notebook*
3. Buscar la opción *Rulers* (dentro del apartado *Code Cell Configuration*).
4. Seleccionar el botón azul `+` para añadir una nueva
5. Escribir 79
6. Deberá aparecer una línea vertical en las líneas de código.

### Sangría

- Sangría de cuatro espacios

Esta recomendación aporta estructura visual al código. En lugar de

```python
selected_cities = [
"madrid", "zaragoza", "valladolid", "valencia", "pontevedra", "lugo",
"sevilla", "huesca", "leon", "zamora",
]
```

aplicar una sangría para entender de manera visual que el código pertenece al mismo bloque.

```python
selected_cities = [
    "madrid", "zaragoza", "valladolid", "valencia", "pontevedra", "lugo",
    "sevilla", "huesca", "leon", "zamora",
]
```

El número de espacios utilizados en la sangría se puede modificar en Jupyter Lab dentro del menú de opciones (`Ctrl + ,`).

1. En el panel izquierdo, seleccionar *Notebook*
2. Buscar la opción *Indentation unit* (dentro del apartado *Code Cell Configuration*).
3. Por defecto, el número de espacios es 4.

### Separar el código

- Utilizar líneas en blanco individuales para dividir el código

Aunque se apliquen las reglas anteriores, el código seguiría siendo difícil de leer si todas las líneas aparecen juntas una detrás de otra.

Incluir líneas en blanco permite separar el código en bloques con sentido:

```python
city = "Paris"
latitude = 48.86
longitude = 2.35

print("City:", city)
print("Latitude:", latitude)

country = "France"
print("Country:", country)
```

### Caracteres especiales

Se recomienda escribir Python utilizando [caracteres ASCII](https://es.wikipedia.org/wiki/ASCII) (evitando tildes, eñes y caracteres especiales en nombres de variables, funciones y archivos).

El objetivo, **mejorar la compatibilidad, portabilidad y legibilidad del código**. Aunque Python 3 admite *Unicode*, utilizar ASCII en los identificadores y rutas puede prevenir problemas de compatibilidad entre sistemas y herramientas.

El inglés es el idioma predominante en la programación, por lo que se recomienda utilizarlo al escribir los comentarios ([Sección de codificación dentro de la PEP 8](https://peps.python.org/pep-0008/#source-file-encoding)). Facilita la colaboración, la consulta de documentación y el mantenimiento del código en proyectos colaborativos.

### Romper las reglas

Incumplir las reglas de estilo no implica que el código Python deje de ejecutarse. De hecho, se recomienda no seguir las recomendaciones cuando:

- Hagan el código menos legible
- El proyecto no siga las recomendaciones. Un ejemplo, la [guía de código de Python de Google](https://google.github.io/styleguide/pyguide.html)
- Compatibilidad con versiones previas de Python que no permitan cumplir la regla

## Comentarios

Un comentario es una línea que el intérprete de Python ignora al ejecutar el código. Para escribir un comentario se utiliza una almohadilla (`#`). Todo lo que aparece después de este símbolo, hasta el final de la línea, se considera un comentario.

Los comentarios ayudan a comprender el código, especialmente cuando se vuelve a consultar después de un tiempo o cuando otras personas necesitan interpretarlo. Sus principales funciones son:

- Explicar el propósito de una sección de código y las decisiones tomadas.
- Aclarar operaciones complejas o poco evidentes.
- Añadir información relevante sobre el código.
- Desactivar temporalmente una línea de código.

````python
# Display a welcome message.
print("Hello world!")
````

No es necesario comentar cada línea. Los comentarios son especialmente útiles cuando:

- existen varias formas de resolver un problema,
- se ha elegido una solución poco evidente o
- puede resultar difícil recordar el razonamiento seguido.

Un buen comentario explica **por qué se hace algo**, en lugar de limitarse a describir lo que ya resulta evidente en el código.

Por ejemplo, el siguiente comentario aporta poca información:

```python
# Add one to the counter.
counter += 1
```

En cambio, este explica el motivo de la operación:

```python
# Include the initial measurement in the total count.
counter += 1
```

Según las recomendaciones de la [PEP 8](https://peps.python.org/pep-0008/), los comentarios deben:

- Ser breves, claros y fáciles de entender.
- Expresar ideas completas, preferiblemente mediante frases.
- Incluir un espacio después de la almohadilla (`# `).
- **Comenzar con mayúscula**, salvo cuando la primera palabra sea un identificador que empiece con minúscula (por ejemplo, el nombre de una variable).
- Terminar con un punto cuando sean frases completas.
- **Mantenerse actualizados** para evitar contradicciones con el código.
- Estar escritos en un idioma consistente. En estos materiales se utilizará el inglés.

Los comentarios que ocupan varias líneas deben mantener la almohadilla y el espacio inicial en cada línea. Cuando contienen varios párrafos, estos se separan mediante una línea que contiene únicamente `#`.

```python
# Calculate the average temperature for the study area.
# Missing values must be excluded to avoid introducing
# bias into the final result.
#
# The result is expressed in degrees Celsius.
```

También pueden escribirse comentarios al final de una línea de código. Se denominan *comentarios en línea* y deben separarse del código mediante al menos dos espacios. Se recomienda utilizarlos únicamente cuando aporten información relevante.

```python
temperature = 25  # Temperature in degrees Celsius.
```

#### Docstrings

Existe un tipo especial de comentarios llamados *docstrings*, que permiten añadir una descripción a los objetos. Es una idea similar al estilo *roxygen2* de R.

Se delimitan por tres comillas dobles (`"""`), y solo se utilizan al comienzo de funciones, clases o módulos.

```python
def add_numbers(a, b):
    """Return the sum of two numbers."""
    return a + b
```

A diferencia de los comentarios, los *docstrings* o cadenas de documentación forman parte del programa y pueden consultarse durante su ejecución.

Se utilizan para documentar el funcionamiento de funciones, clases y módulos. Se explicarán con más detalle al abordar este tipo de estructuras. Sus reglas de estilo se recogen en la [PEP 257](https://peps.python.org/pep-0257/).
