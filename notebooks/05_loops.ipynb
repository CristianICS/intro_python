# Bucles en Python

Este capítulo explica los bucles (*loops*), mostrando cómo repetir operaciones.

## Repetición de instrucciones: `for` y `while`

Un **bucle** ejecuta un bloque de instrucciones varias veces. Python utiliza principalmente dos estructuras:

- **`for`**: recorre los elementos de una secuencia o de otro objeto iterable.
- **`while`**: repite un bloque mientras una condición sea verdadera. Resulta adecuado cuando el número de repeticiones depende de una condición o de una entrada del teclado.

En ambos casos, las instrucciones que pertenecen al bucle aparecen con una **sangría** (se recomiendan cuatro espacios) y la cabecera termina en dos puntos (`:`).

## El bucle `for`

### Recorrer una lista

La sintaxis habitual es:

```python
for elemento in coleccion:
    # Operaciones
``` 

En cada iteración, `elemento` recibe el siguiente valor de la colección. La lista original no se modifica por el hecho de recorrerla.

En este ejemplo se recorren los nombres de varias estaciones meteorológicas.

````python
stations = ["Zaragoza", "Huesca", "Teruel"]

for station in stations:
    print("Weather station:", station)

print("All stations processed.")
````

El mensaje situado fuera del bloque con sangría se ejecuta **una sola vez**, cuando termina el bucle. Si la lista está vacía, el cuerpo o la operación dentro del `for` no se ejecuta.

Otro ejemplo de cómo recorrer listas, ahora limpiando y convirtiendo datos de texto a formato numérico.

````python
elevations = "200, 300, 400 , 500 ,  600"

for elevation in elevations.split(", "):
    elevation_int = int(elevation)
    print(f"Convert '{elevation}' (str) to {elevation_int} (int)")
````

### Recorrer valores numéricos con `range()`

`range(inicio, fin, paso)` genera una sucesión de números. **El valor `fin` no se incluye**. Si se omite `inicio`, se empieza en 0; si se omite `paso`, se avanza de uno en uno.

En el cuaderno introductorio sobre *listas* se vio como utilizar `range()` para crear una lista de números. En este caso, la salida de la función (objeto de tipo `range`) se puede utilizar directamente en un bucle.

El siguiente ejemplo combina el uso de `range()` y del bucle `for`, representando el identificador de cada celda de una cuadrícula.

````python
for cell_id in range(1, 6):
    print("Grid cell:", cell_id)
````

### Realizar cálculos sobre observaciones

Un bucle puede utilizarse para calcular una operación sobre los elementos de una lista. En estos casos, es recomendable utilizar una **variable acumuladora**: una variable definida **fuera del bucle** que guardará el resultado a medida que se calculan los valores de la lista.

Es importante que se inicialice antes del bucle, asegurándose de que no existe anteriormente con un valor no deseado.


En el siguiente ejemplo, cada número representa la precipitación diaria, en milímetros.

````python
rainfall_mm = [2.5, 0.0, 4.0, 1.5]
total_rainfall_mm = 0.0

for daily_rainfall in rainfall_mm:
    total_rainfall_mm += daily_rainfall
    day = rainfall_mm.index(daily_rainfall) + 1
    print("Day", day, "- accumulated rainfall (mm):", total_rainfall_mm)

mean_rainfall = total_rainfall_mm / len(rainfall_mm)
print("Mean rainfall (mm):", mean_rainfall)
````

El código anterior daría un error si la lista estuviese vacía, es decir, si no hubiera ningún dato de precipitación. Esto puede evitarse añadiendo una condición que evalúe el número de elementos en la lista.

**Error intencionado: `ZeroDivisionError`.** En la siguiente celda, la lista está vacía y `len(rainfall_mm)` vale `0`, por lo que la media divide entre cero.

````python
rainfall_mm = []
total_rainfall_mm = 0.0

for daily_rainfall in rainfall_mm:
    total_rainfall_mm += daily_rainfall

mean_rainfall = total_rainfall_mm / len(rainfall_mm)
````

````python
rainfall_mm = []
total_rainfall_mm = 0.0

for daily_rainfall in rainfall_mm:
    total_rainfall_mm += daily_rainfall

if len(rainfall_mm) != 0:
    mean_rainfall = total_rainfall_mm / len(rainfall_mm)
    print("Mean rainfall (mm):", mean_rainfall)
else:
    print("There is no valid data.")
````

#### Asignación aumentada

En el ejemplo anterior, la variable `total_rainfall_mm` se inicia con un valor 0, que irá aumentando a medida que vayan sumándose los datos de precipitación.

En Python, si en lugar del operador de asignación `=` se utiliza `+=`, se interpreta como:

```python
total_rainfall_mm = total_rainfall_mm + daily_rainfall
```

Además del operador `+`, se pueden utilizar el resto de operadores matemáticos junto con el símbolo de asignación `=`, denominados **operadores de asignación aumentada**:

| Operador | Equivalente a | Significado |
|---|---|---|
| `x += y` | `x = x + y` | suma |
| `x -= y` | `x = x - y` | resta |
| `x *= y` | `x = x * y` | multiplicación |
| `x /= y` | `x = x / y` | división |
| `x //= y` | `x = x // y` | división entera (sin decimales) |
| `x %= y` | `x = x % y` | resto |
| `x **= y` | `x = x ** y` | potenciación (exponente) |

````python
x = 10
print(f"x = {x}")

x *= 3
print(f"x *= 3 -> {x}")

x /= 4
print(f"x /= 4 -> {x}")

x //= 2
print(f"x //= 2 -> {x}")

x %= 2
print(f"x %= 2 -> {x}")

x **= 3
print(f"x **= 3 -> {x}")
````

Importante:

- **El operador debe colocarse justo antes de `=`.** Escribir `x =+ 1` no devuelve un error, se interpreta como `x = +1`, asignando a la variable `x` un valor positivo `1`.
- Este tipo de operaciones también funcionan con textos y listas:
    - `+=` concatena textos y extiende listas
    - `*=` repite elementos

````python
message = "F"
message *= 6
print(message)

rivers = ["ebro"]
rivers += ["gallego"]
print(rivers)
````

### Recorrer una lista de tuplas

La información espacial suele agruparse en colecciones de elementos que contienen varios atributos. Por ejemplo:

```python
coordinates = [(45, 24), (50, 21), (46, 25)]
```

La lista `coordinates` contiene tuplas, cada una de ellas con sus valores de latitud y longitud.

Un bucle `for` permite recorrer estos registros y realizar operaciones sobre sus atributos sin necesidad de acceder manualmente a cada posición de la lista.

Utilizando la lista `coordinates`, el bucle puede acceder a los valores latitud y longitud mediante el **desempaquetado de tuplas** (*tuple unpacking*):

````python
coordinates = [(45, 24), (50, 21), (46, 25)]

for lat, long in coordinates:
    print(f"Latitude {lat}, Longitude {long}")
````

En cada iteración, Python asigna automáticamente el primer elemento de la tupla a `lat` y el segundo a `long`.

Este procedimiento evita tener que acceder a cada valor mediante su índice, como `coordinate[0]` y `coordinate[1]` en un bucle `for coordinate in coordinates:`, facilitando la lectura del código.

**Importante:** el número de variables utilizadas en el desempaquetado debe coincidir con el número de elementos de cada tupla. De lo contrario, se recibirá un error del tipo `ValueError`.

**Error intencionado: `ValueError`.** En la siguiente celda, cada tupla tiene dos valores, pero se desempaquetan en tres variables.

````python
coordinates = [(45, 24), (50, 21), (46, 25)]

for lat, long, z in coordinates:
    print(f"Latitude {lat}, Longitude {long}")
````

#### Enumerate

Volviendo al cálculo de operaciones dentro de bucles `for`, en el ejemplo anterior con las precipitaciones se utilizaba `index()` para obtener el día de la medición. Esta opción no es la mejor, ya que los valores repetidos siempre obtendrán el índice del primer valor.

En Python, la función `enumerate()` permite crear una lista de tuplas a partir de los valores de una lista. Cada una de las tuplas contiene el valor de la lista junto a su índice de posición en la misma.

````python
rainfall_mm = [2.5, 0.0, 4.0]

print(list(enumerate(rainfall_mm)))
````

En el ejemplo anterior, el primer valor de la lista es `(0, 2.5)`, siendo 0 la posición del valor `2.5` en la lista `rainfall_mm`.

El resultado de `enumerate()` no es una lista, sino un objeto de tipo `enumerate`. Se ha convertido a lista para poder mostrarse en pantalla, pero no es necesario cuando se utiliza dentro de un bucle `for`, como en el siguiente ejemplo.

````python
rainfall_mm = [2.5, 0.0, 4.0, 1.5]

for day, daily_rainfall in enumerate(rainfall_mm, start=1):
    print("Day", day, "- rain (mm):", daily_rainfall)
````

La lista de tuplas obtenida con `enumerate()` se ha dividido en dos variables dentro del bucle, una con el día de la medición y otra con su valor.

Como el día se relaciona con la posición del dato en la variable, se ha especificado el argumento `start=1`, haciendo que los índices comiencen con el valor `1` en lugar de `0`.

`start=1` solo cambia el número asignado; el primer elemento de la lista sigue siendo el primero en recorrerse.

````python
print(list(enumerate(rainfall_mm, start=1)))
````

#### zip

En ocasiones los datos llegan en **listas paralelas**: por ejemplo, una lista con los nombres de las estaciones y otra con sus elevaciones, donde la posición 0 de ambas corresponde a la misma estación.

La función `zip()` empareja los elementos de dos o más listas según su posición y devuelve tuplas. Al igual que `enumerate()`, el resultado es un objeto de tipo `zip` que se puede convertir a lista para mostrarlo, aunque no es necesario dentro de un bucle `for`.

````python
names = ["Zaragoza", "Huesca", "Teruel"]
elevations = [200, 488, 915]

print(list(zip(names, elevations)))
````

Dentro de un bucle, las tuplas se desempaquetan igual que con `enumerate()` o con una lista de tuplas: una variable por cada lista.

````python
for name, elevation in zip(names, elevations):
    print(f"{name}: {elevation} m")
````

Se pueden combinar más de dos listas, y también `zip()` con `enumerate()` (en ese caso, hay que desempaquetar con paréntesis la tupla que viene de `zip()`):

````python
latitudes = [41.65, 42.14, 40.34]
longitudes = [-0.88, -0.41, -1.11]

for name, lat, long in zip(names, latitudes, longitudes):
    print(f"{name} ({lat}, {long})")

for number, (name, elevation) in enumerate(zip(names, elevations), start=1):
    print(f"{number} - {name}: {elevation} m")
````

**Importante:** si las listas tienen distinta longitud, `zip()` se detiene en la más corta y descarta, **sin avisar**, los elementos sobrantes. Conviene comprobar las longitudes antes de emparejar datos.

````python
names = ["Zaragoza", "Huesca", "Teruel"]
elevations = [200, 488]

print(list(zip(names, elevations)))
````

## Comprensión de listas (*list comprehensions*)

Las listas y los bucles *for* pueden combinarse, condensando el bucle en **una sola línea** y dando como resultado una lista.

Es una alternativa al bucle `for` cuando se desea transformar o filtrar datos.

Su sintaxis básica es:

```python
new_list = [expression for element in collection]
```

- `expression`: operación que se aplica a cada elemento, de la que se deriva el resultado final.
- `element`: variable que representa cada valor de la colección.
- `collection`: colección de datos que se recorre.

La comprensión devuelve una **lista nueva**, sin modificar la original.

### Transformación de datos

Por ejemplo, se pueden convertir las elevaciones de varios puntos de muestreo de metros a kilómetros.

````python
elevations_m = [450, 780, 1200, 1650]

# Using a for loop
elevations_km = []

for elevation in elevations_m:
    elevations_km.append(elevation / 1000)

print(elevations_km)

# Using a list comprehension
elevations_km = [elevation / 1000 for elevation in elevations_m]

print(elevations_km)
````

Ambos procedimientos producen el mismo resultado: `[0.45, 0.78, 1.2, 1.65]`.

La comprensión de listas reúne en una sola línea la creación de la lista, el recorrido de sus elementos y la operación aplicada.

### Eliminación de espacios en cadenas de texto

Al importar coordenadas, nombres de estaciones o atributos de un fichero de texto, pueden aparecer espacios innecesarios.

El método `.strip()` elimina los espacios en blanco situados **al principio y al final** de una cadena, pero conserva los espacios interiores.

Una comprensión de listas permite aplicar este método a todos los elementos de una colección.

````python
coordinates = "  station_1 ,  station_2  ,   station_3, station_4  "

clean_coordinates = [coord.strip() for coord in coordinates.split(",")]

print(clean_coordinates)
````

### Filtrado de elementos mediante `if`

También se puede añadir una condición `if` para seleccionar únicamente los elementos que cumplen un criterio.

La sintaxis es:

```python
new_list = [expression for element in collection if condition]
```

Por ejemplo, en un modelo digital de elevaciones se pueden seleccionar las altitudes superiores a 1.000 metros.

La condición se evalúa para cada elemento y solo se incorporan a la nueva lista aquellos que la cumplen.

````python
elevations_m = [450, 780, 1200, 1650, 920, 2100]

high_elevations = [elevation for elevation in elevations_m if elevation > 1000]

print(high_elevations)
````

En ocasiones, la línea que ejecuta la comprensión es demasiado grande. En este caso, y para cumplir con la regla de estilo en la PEP 8, se deberá romper la línea en varias partes.

De esta forma, se pierde la ventaja de realizar la operación en una línea, pero se sigue optimizando la operación, eliminando por lo menos las dos líneas necesarias para crear la nueva lista y utilizar `append()` para incluir los nuevos elementos.

````python
elevations_m = [450, 780, 1200, 1650, 920, 2100, 340, 1890, 1010, 2500]

feet_above_1000m = [round(elevation * 3.28084) for elevation in elevations_m if elevation > 1000]
````

El ejemplo anterior convierte las elevaciones de metros a pies, pero solo si son superiores a 1 km. La línea de la comprensión supera los 79 caracteres, por lo que se debería romper de la siguiente forma:

````python
elevations_m = [450, 780, 1200, 1650, 920, 2100, 340, 1890, 1010, 2500]

feet_above_1000m = [
    round(elevation * 3.28084)
    for elevation in elevations_m
    if elevation > 1000
]

print(feet_above_1000m)
````

Dentro de los corchetes `[]`, Python permite romper las líneas libremente. La forma utilizada en la celda anterior es la recomendada ya que cada una de las líneas se relaciona con las partes de la comprensión: 

- la expresión
- el bucle
- el filtro

## El bucle `while`

Un bucle `while` **comprueba la condición antes de cada iteración**. Mientras la condición sea `True`, ejecuta el bloque con sangría. Cuando pasa a ser `False`, continúa con la siguiente instrucción exterior fuera del bloque. Si la condición es falsa desde el principio, el bloque no se ejecuta ninguna vez.

El siguiente ejemplo avanza un kilómetro en cada iteración hasta alcanzar 5 km.

````python
distance_km = 0

while distance_km < 5:
    print("Distance covered (km):", distance_km)
    distance_km += 1

print("Target distance reached (km):", distance_km)
````

La variable `distance_km` comienza en 0. **Su actualización es imprescindible para que cambie la condición de salida**. En este caso se muestran las distancias 0, 1, 2, 3 y 4, y el bucle termina cuando se alcanza 5.

Un bucle `while` **no necesita comenzar con una condición verdadera**: también es válido que no llegue a ejecutarse. En el ejemplo anterior, esto sucedería iniciando la variable `distance_km` con un valor 5.

### Procesar una lista con `pop()`

Una de las utilidades del bucle `while` es permitir el procesado de los valores de una lista hasta que se quede vacía.

Por ejemplo, si se han descargado una serie de imágenes y guardado su nombre en una lista, se podrán procesar dentro del bucle `while` y este parará cuando la lista esté vacía. Es decir, cuando ya se hayan procesado todas las imágenes.

````python
images = [f"file{i}.tif" for i in range(1, 11)]

while images:
    image = images.pop()
    print(f"Processing image {image}")

print("Done.")
````

Para poder realizar este proceso con un bucle `while`, es necesario utilizar el método `pop()`, visto en el cuaderno de listas.

Permite extraer el último elemento de la lista e incluirlo en una nueva variable. Al eliminar el elemento, cada vez que se ejecuta el bucle queda una imagen menos en la lista.

## Bucles infinitos y errores frecuentes

Un **bucle infinito accidental** ocurre cuando su condición sigue siendo verdadera y no existe una salida efectiva. En Jupyter se puede detener desde **Kernel -> Interrupt Kernel**. Reiniciar el núcleo o *kernel* elimina las variables de la memoria.

Los bucles infinitos se suelen producir por falta de actualización de la condición o por una actualización en sentido incorrecto.

### Falta de actualización

El fragmento siguiente es un **código incorrecto y no debe ejecutarse**: `distance_km` nunca cambia.

```python
# Infinite loop
distance_km = 0
while distance_km < 3:
    print(distance_km)
```

La versión correcta incrementa el contador dentro del bucle.

````python
distance_km = 0

while distance_km < 3:
    print("Distance (km):", distance_km)
    distance_km += 1
````

### Actualización en sentido incorrecto

Si la condición es `distance_km < 3` y el valor inicial es 0, disminuirlo con `distance_km -= 1` aleja el contador del límite y el bucle tampoco termina.

### Otros errores habituales

- **Límite de `range()`:** `range(1, 5)` produce 1, 2, 3 y 4, no 5.
- **Sangría:** una línea sin sangría está fuera del cuerpo del bucle.

## Comparación entre `for` y `while`

El siguiente código muestra las celdas 1 a 4. Con `for` se define directamente la secuencia; con `while` es necesario controlar el contador.

````python
for cell_id in range(1, 5):
    print("FOR cell:", cell_id)

cell_id = 1
while cell_id < 5:
    print("WHILE cell:", cell_id)
    cell_id += 1
````

## Resumen

- `for` recorre los elementos de una colección; `range()` define sucesiones de enteros con el límite superior excluido.
- `enumerate()` asocia a cada elemento su posición; el argumento `start` cambia el número inicial.
- Las comprensiones de listas crean una lista nueva en una sola expresión y pueden incluir un filtro `if`.
- `while` ejecuta instrucciones mientras una condición sea verdadera.
- `pop()` extrae elementos de una lista y permite iterarla en un bucle `while`.
- `zip()` empareja listas paralelas; se detiene en la más corta.
- Al revisar un bucle conviene comprobar: **condición, actualización, sangrado y salida**.
