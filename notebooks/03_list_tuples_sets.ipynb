## Listas

Una **lista** agrupa varios valores en un orden determinado. Se escribe entre corchetes (`[]`) y sus elementos se separan mediante comas. Puede contener texto, números u otros tipos de datos. Se guarda en una variable igual que cualquier otro valor.

En este ejemplo, la lista contiene tres ciudades:

````python
cities = ["barcelona", "zaragoza", "madrid"]
print(cities)
````

### Crear y nombrar listas

Se recomienda utilizar nombres descriptivos en plural para las variables que contienen listas, por ejemplo, `cities` o `land_cover_types`. Los elementos se escriben entre corchetes y separados por comas. No es obligatorio que todos sean del mismo tipo, aunque utilizar valores relacionados facilita su interpretación.

````python
land_cover_types = ["shrub", "forest", "peatland"]
print(land_cover_types)
````

### Acceder a un elemento por su posición

Cada elemento tiene un **índice**, un número que indica su posición dentro de la lista. En Python, el primer índice es `0`, el segundo es `1` y el tercero es `2`.

A diferencia de R, cuyos índices empiezan en `1`, Python comienza en `0`.

Para acceder a un elemento se escribe el nombre de la lista seguido del índice entre corchetes: `cities[0]`.

````python
cities = ["barcelona", "zaragoza", "madrid"]
city = cities[0]

print(city.title())
````

El índice `0` corresponde al primer elemento. Para consultar el segundo, se utiliza `cities[1]`. El método `.title()`, presentado en el cuaderno anterior, permite mostrar el nombre con mayúscula inicial sin modificar la lista.

````python
cities = ["barcelona", "zaragoza", "madrid"]
city = cities[1]
print(city.title())
````

### Acceder a elementos desde el final

El índice `-1` permite seleccionar el último elemento, independientemente de cuántos contenga la lista. El índice `-2` selecciona el penúltimo.

````python
cities = ["barcelona", "zaragoza", "madrid"]
city = cities[-1]
print(city.title())
````

El índice `-2` corresponde al penúltimo elemento:

````python
cities = ["barcelona", "zaragoza", "madrid"]
city = cities[-2]
print(city.title())
````

### Modificar y consultar listas

Las listas son **mutables**: sus elementos se pueden sustituir, añadir o eliminar después de crear la lista.

#### Sustituir un elemento

Para cambiar un elemento se indica su índice a la izquierda de `=` y el nuevo valor a la derecha. En el ejemplo se sustituye el tercer elemento (índice `2`).

````python
cities = ["Madrid", "Zaragoza", "Belchite"]

cities[2] = "Barcelona"
print(cities)
````

#### Buscar la posición de un elemento

El método `.index()` devuelve el índice de la **primera aparición** de un valor, incluido como argumento.

> Los valores aceptados por una función se denominan **argumentos**.

La búsqueda distingue entre mayúsculas y minúsculas. El método `.index()` consulta la lista, pero **no la modifica**.

````python
provinces = ["aragon", "asturias", "galicia"]

print(provinces.index("asturias"))
````

El resultado `1` significa que `"asturias"` ocupa la segunda posición.

#### Comprender un error: `ValueError`

El texto `"Asturias"` (con mayúscula) no coincide con `"asturias"`. Cuando `.index()` no encuentra el valor, produce `ValueError`, no `IndexError`.

````python
provinces = ["aragon", "asturias", "galicia"]
print(provinces.index("Asturias"))
````

#### Comprobar si un elemento está presente

El operador `in` comprueba si un valor se encuentra en la lista y devuelve `True` o `False`. A diferencia de `.index()`, no produce un error cuando el valor no aparece.

````python
provinces = ["aragon", "asturias", "galicia"]

print("aragon" in provinces)
print("Aragón" in provinces)
````

#### Añadir un elemento al final

El método `.append()` añade un elemento al final de una lista existente. **Cambia** la lista original.

````python
rivers = ["ebro", "tajo", "duero"]

rivers.append("guadalquivir")
print(rivers[-1])
````

`.append()` modifica la lista original sin devolver ningún valor. Es por ello que, cuando se imprime el resultado de la operación, aparece un valor `None`.

`None` representa la ausencia de un valor. Por este motivo, la lista modificada se comprueba llamando a la variable original.

````python
rivers = ["ebro", "tajo", "duero"]
result = rivers.append("guadalquivir")
print(result)
print(rivers)
````

#### Insertar un elemento en una posición

El método `.insert(indice, valor)` inserta un valor en la posición indicada. Los elementos que estaban en esa posición y las siguientes se desplazan hacia la derecha.

````python
rivers = ["ebro", "tajo", "duero"]
rivers.insert(1, "guadalquivir")

print(rivers)
````

#### Comprender un error: `TypeError`

`.insert()` espera primero un índice entero y después el valor. Al invertir ambos argumentos, Python no puede utilizar el texto como índice.

````python
rivers = ["ebro", "tajo", "duero"]
rivers.insert("guadalquivir", 1)
````

### Ordenar listas

El método `.sort()` ordena los elementos de la **lista original**. En una lista de textos, el orden se determina por los caracteres; en una lista de números, por sus valores. Para evitar resultados inesperados, los ejemplos utilizan listas cuyos elementos son del mismo tipo.

````python
mountains = ["teide", "aneto", "everest"]

mountains.sort()
print(mountains)
````

Con `.sort(reverse=True)` se obtiene el orden descendente, por ejemplo de la Z a la A o del valor mayor al menor.

````python
mountains = ["teide", "aneto", "everest"]

mountains.sort(reverse=True)
print(mountains)
````

En el método `sort()` se incluye el nombre del argumento además de su valor. Al utilizar los métodos `append()` o `index()` no era necesario.

El motivo es que la función `sort()` tiene un argumento que va antes que `reverse`. Los argumentos de `sort()` deben indicarse siempre con su nombre (`reverse=True`). Escribir `sort(True)` produce un `TypeError`.

````python
help(list.sort)
````

`.sort()` cambia la lista y no devuelve una lista nueva. En cambio, la función `sorted()` devuelve una **lista nueva** con los elementos ordenados y conserva la lista original.

````python
mountains = ["teide", "aneto", "everest"]

print(f"Sorted list: {sorted(mountains, reverse=True)}")
print(f"Original list: {mountains}")
````

El método `.reverse()` invierte el orden actual de los elementos **sin ordenarlos**. Esto es diferente de ordenar en orden descendente: `sorted(..., reverse=True)` sí ordena los valores.

````python
mountains = ["teide", "aneto", "everest", "mulhacen"]

print(f"Descending alphabetical order: {sorted(mountains, reverse=True)}")
mountains.reverse()
print(f"Reversed original order: {mountains}")
````

Las listas numéricas también pueden ordenarse. El ejemplo siguiente compara los órdenes ascendente y descendente.

````python
heights = [1200, 800, 750, 20]

print(sorted(heights))
print(sorted(heights, reverse=True))
````

### Eliminar elementos

Al eliminar un elemento se **modifica la lista original**. No podrá recuperarse a no ser que se guarde una copia antes de realizar los cambios.

#### Eliminar por índice con `del`

La instrucción `del` elimina el elemento situado en el índice indicado. `del pending_images[0]` elimina el primer elemento.

````python
pending_images = ["S2A_20160519", "S2C_20170604", "S2B_20221104"]
del pending_images[0]

print(pending_images)
````

#### Eliminar por valor con `.remove()`

El método `.remove(valor)` elimina la primera aparición del valor indicado. Si no está presente, produce un error `ValueError`.

````python
pending_images = ["S2A_20160519", "S2C_20170604", "S2B_20221104"]
pending_images.remove("S2A_20160519")

print(pending_images)
````

Si el mismo valor aparece varias veces, `.remove()` elimina solo la primera coincidencia.

````python
aspects = ["N", "S", "N", "N", "E", "NE"]
aspects.remove("N")
print(aspects)
````

### Extraer un elemento con `.pop()`

El método `.pop()` **elimina y devuelve** el último elemento. Puede guardarse en otra variable para utilizarlo después. En el ejemplo se simula la selección de una imagen pendiente de descarga en una lista de imágenes.

````python
images_to_download = ["S2C_001", "S2C_002", "S2C_003"]
image = images_to_download.pop()
print(f"Selected image: {image}")
print(f"Remaining images: {images_to_download}")
````

Sin indicar un índice, `.pop()` selecciona el último elemento. También permite incluir un índice para eliminar el elemento según su posición.

````python
images_to_download = ["S2C_001", "S2C_002", "S2C_003"]
image = images_to_download.pop(0)
print(f"Selected image: {image}")
print(f"Remaining images: {images_to_download}")
````

### Seleccionar un fragmento (*slicing*)

Un **segmento** es una selección de elementos consecutivos. Se obtiene con `lista[inicio:fin]`: incluye el índice `inicio`, pero **excluye** el índice `fin`. Así, `rivers[0:3]` selecciona los tres primeros elementos y devuelve una lista nueva.

````python
rivers = ["ebro", "huerva", "gallego", "aragon", "aguasvivas"]

first_three = rivers[0:3]
print(first_three)
````

Si se omite el índice inicial, la selección comienza en el primer elemento. `rivers[:3]` equivale a `rivers[0:3]`.

````python
rivers = ["ebro", "huerva", "gallego", "aragon", "aguasvivas"]
print(rivers[:3])
````

La selección de un segmento crea una lista nueva y no altera la lista original.

````python
rivers = ["ebro", "huerva", "gallego", "aragon", "aguasvivas"]

first_three = rivers[0:3]
print(first_three)
print(rivers)
````

También se puede seleccionar un fragmento intermedio. `rivers[1:4]` incluye los índices `1`, `2` y `3`.

````python
rivers = ["ebro", "huerva", "gallego", "aragon", "aguasvivas"]
print(rivers[1:4])
````

Al omitir el índice final, se seleccionan los elementos desde la posición indicada hasta el final.

````python
rivers = ["ebro", "huerva", "gallego", "aragon", "aguasvivas"]
print(rivers[2:])
````

### Copiar una lista

Asignar una lista a otra variable **no crea una copia**: las dos variables se refieren a la misma lista. Si se modifica a través de cualquiera de los nombres, el cambio se observa desde ambos.

````python
cities = ["zaragoza", "madrid", "barcelona", "valencia"]
another_cities = cities

another_cities.sort()

print(cities)
print(another_cities)
````

Para crear una copia independiente de una lista sencilla, se puede utilizar `[:]`, que selecciona todos los elementos. Así, ordenar la copia no cambia la lista original.

````python
cities = ["zaragoza", "madrid", "barcelona", "valencia"]
another_cities = cities[:]

another_cities.sort()

print(cities)
print(another_cities)
````

Ahora `another_cities` y `cities` son listas distintas. Esta forma de copiar funciona de manera independiente en los ejemplos con texto y números. Si una lista contiene otras listas, `[:]` **no copia por separado esas listas internas**; ese caso se tratará más adelante.

### Listas de números

Las listas pueden contener números, y estas pueden operarse con funciones básicas de Python.

#### Crear secuencias con `range()`

La función `range(inicio, fin, paso)` representa una secuencia de números enteros. Incluye `inicio`, excluye `fin` y avanza según `paso`. Por ejemplo, `range(100, 1000, 50)` representa 100, 150, 200… hasta un valor menor que 1000.

````python
contour_altitudes = range(100, 1000, 50)
print(contour_altitudes)
````

`range()` no crea directamente una lista. Representa la secuencia de números sin almacenar todos sus elementos como una lista. Más adelante se utilizará en los bucles.

````python
contour_altitudes = range(100, 1000, 50)
print(type(contour_altitudes))
````

La función `list()` permite convertir un `range` en una lista para mostrar todos sus valores. Al igual que `str()` o `int()`, recibe un valor entre paréntesis, pero el tipo de resultado es diferente.

````python
contour_altitudes = list(range(100, 1000, 50))
print(contour_altitudes)
````

#### Obtener mínimos, máximos y sumas

En listas numéricas, `min()` devuelve el valor mínimo, `max()` el máximo y `sum()` la suma total.

````python
rainfall = [12, 5, 0, 18, 7, 3, 10]

print("Minimum rainfall:", min(rainfall), "mm")
print("Maximum rainfall:", max(rainfall), "mm")
print("Total rainfall:", sum(rainfall), "mm")
````

En este ejemplo, cada número representa la precipitación en milímetros durante un día. `sum()` calcula la precipitación acumulada de los días registrados; `min()` y `max()` identifican los valores extremos.

## Cadenas de texto como secuencias

En el cuaderno anterior se introdujo el tipo `str`. Una cadena de texto también es una **secuencia**: sus caracteres tienen posiciones y pueden seleccionarse mediante índices y segmentos.

Sin embargo, a diferencia de las listas, las cadenas de texto son **inmutables**: sus caracteres no se pueden sustituir directamente.

### Índices, segmentos y pertenencia en textos

Se puede seleccionar el primer carácter con `[0]`:

````python
coordinate = "41.6488,-0.8891"
print(coordinate[0])
````

Con `[:7]` se seleccionan los siete primeros caracteres de la cadena:

````python
coordinate = "41.6488,-0.8891"
print(coordinate[:7])
````

El operador `in` permite comprobar si una coma aparece en el texto:

````python
coordinate = "41.6488,-0.8891"
print("," in coordinate)
````

La función `len()` devuelve el número de caracteres de la cadena, incluidos signos y espacios:

````python
coordinate = "41.6488,-0.8891"
print(len(coordinate))
````

### Buscar fragmentos de texto

Una **subcadena** es una secuencia de caracteres dentro de otra cadena de texto. El operador `in` comprueba su presencia y devuelve `True` o `False`.

````python
filename = "sentinel2_zaragoza_2024.tif"
sentinel_present = "sentinel2" in filename
print(sentinel_present)
````

El método `.find()` devuelve el índice donde empieza la **primera** coincidencia. Si el texto buscado no aparece, devuelve `-1`, en lugar de producir un error.

````python
filename = "sentinel2_zaragoza_2024.tif"
city_index = filename.find("zaragoza")
print(city_index)

city = filename[city_index:city_index + len("zaragoza")]
print(city.title())
````

Si el texto se repite, `.find()` devuelve únicamente la posición de la primera coincidencia.

````python
filename = "sentinel2_zaragoza_sentinel2_2024.tif"
sentinel_index = filename.find("sentinel2")
print(sentinel_index)
````

El método `.rfind()` devuelve la posición de la **última** coincidencia; igual que `.find()`, devuelve `-1` cuando no encuentra el texto.

````python
filename = "sentinel2_zaragoza_sentinel2_2024.tif"
last_sentinel_index = filename.rfind("sentinel2")
print(last_sentinel_index)
````

### Sustituir fragmentos con `.replace()`

El método `.replace(texto_anterior, texto_nuevo)` devuelve una **cadena nueva** con las sustituciones. No modifica la cadena original: para conservar el resultado, se asigna a una variable.

````python
filename = "landcover_forest_2024.tif"
filename = filename.replace("forest", "shrubland")
print(filename)
````

### Contar apariciones con `.count()`

El método `.count()` devuelve cuántas veces aparece una subcadena dentro de un texto.

````python
land_cover = "forest, shrubland, Forest, grassland, forest"
forest_count = land_cover.count("forest")
print(forest_count)

forest_count = land_cover.lower().count("forest")
print(forest_count)
````

Se puede consultar el resto de métodos de texto en la [documentación oficial de Python](https://docs.python.org/3/library/stdtypes.html#string-methods). No es necesario memorizarlos: lo importante es reconocer las operaciones disponibles.

### Convertir textos en listas

El método `.split(separador)` divide una cadena en partes y devuelve una lista. Por ejemplo, extraer la localización de una imagen contenida en su nombre de archivo (ejemplo del ejercicio 9) se puede hacer más fácilmente:

````python
filename = "sentinel2_zaragoza_2024.tif"
filename_parts = filename.split("_")

print(filename_parts)
print(filename_parts[1])
````

También se puede usar para separar las dos coordenadas guardadas en un mismo texto.

````python
coordinate = "41.6488, -0.8891"
coordinate_parts = coordinate.split(",")
print(coordinate_parts)
````

Al dividir por `","`, el espacio situado después de la coma se conserva al comienzo del segundo elemento. Se puede eliminar con `.strip()`, método ya presentado en el cuaderno anterior.

````python
coordinate = "41.6488, -0.8891"
coordinate_parts = coordinate.split(",")
print(coordinate_parts[1].strip())
````

### Convertir listas en textos
 
El método `.join()` une varios textos de una lista en una sola cadena.

El separador con el que se unen los elementos se escribe antes de `.join()`. Por ejemplo, `", ".join(cities)` une los nombres de ciudades con comas y espacios.

````python
cities = ["zaragoza", "madrid", "barcelona", "valencia"]
print(", ".join(cities).title())
````

También permite crear un identificador de texto separando sus partes mediante guiones bajos (`_`).

````python
location = ["spain", "aragon", "zaragoza"]
identifier = "_".join(location)

print(identifier)
````

`.join()` no cambia la lista. Todos los elementos que se unen deben ser cadenas de texto (`str`); si hay números, primero deben convertirse.

## Tuplas

Una **tupla** es una secuencia ordenada de elementos, parecida a una lista, pero **inmutable**: después de crearla no se pueden sustituir, añadir ni eliminar sus elementos directamente. Resulta útil para agrupar valores que deben mantenerse asociados, como la longitud y la latitud de un punto.

### Crear tuplas y acceder a sus elementos

Una tupla se escribe habitualmente con valores separados por comas entre paréntesis `()`. Los índices comienzan en `0`, igual que en las listas.

En el ejemplo, el primer valor representa la **longitud** y el segundo la **latitud**, en grados decimales, del entorno de Zaragoza (sistema WGS 84, EPSG:4326).

````python
coordinates = (-0.8891, 41.6488)

print(f"Longitud: {coordinates[0]}")
print(f"Latitud: {coordinates[1]}")
````

**Error intencionado: `TypeError`.** Una tupla no admite la sustitución de elementos mediante un índice. Por eso la siguiente asignación produce un error.

````python
coordinates = (-0.8891, 41.6488)
coordinates[0] = -0.888
````

Puede asignarse una **tupla nueva** a la misma variable. Para crear una tupla con un solo elemento, se necesita una coma final: `(25830,)`. Sin la coma, `(25830)` es simplemente un número entre paréntesis.

````python
coordinates = (-0.8891, 41.6488)

# Assign a new tuple
coordinates = (-0.9000, 41.6500)
print(coordinates)

epsg_code = (25830,)
print(type(epsg_code))
````

### Tuplas con listas en su interior

La tupla impide sustituir sus elementos, pero un elemento puede ser, a su vez, una lista **mutable**. En ese caso, el contenido de la lista interna sí puede modificarse.

El siguiente ejemplo combina ambos comportamientos. Primero se modifica un número de la lista interna y después se intenta sustituir toda la lista, lo que produce un **`TypeError` intencionado**.

````python
location = ("zaragoza", [-0.8891, 41.6488])
location[1][0] = -0.9000
print(location)

location[1] = [-0.9000, 41.6500]
````

## Conjuntos (`set`)

Un **conjunto** es una colección de valores **sin duplicados** y **sin un orden fijo**. No permite acceder a sus elementos mediante índices.

Los conjuntos se crean a partir de listas, utilizando `set(lista)`. Sus elementos deben ser **inmutables**, como las cadenas de texto o las tuplas (valores aptos para identificarse, ya que no cambian durante su permanencia en el conjunto). Por el mismo motivo, las listas **no pueden ser** elementos de un conjunto.

Es útil, por ejemplo, para identificar clases de cobertura del suelo distintas dentro de varias observaciones o si existen coordenadas únicas.


````python
land_cover = ["forest", "forest", "water", "urban", "forest", "cropland"]
land_cover_set = set(land_cover)

print(land_cover_set)
````

````python
coordinates = [(-0.8891, 41.6488), (-0.8891, 41.6488)]

print(set(coordinates))
````

`set()` elimina los duplicados. También se puede crear un conjunto directamente entre llaves `{}`. Como no existe un orden fijo, los elementos pueden aparecer en posiciones diferentes al mostrar el conjunto. Un conjunto vacío se crea con `set()`, **no** con `{}`.

````python
land_cover_set = {"forest", "water", "urban", "cropland"}
print(sorted(land_cover_set))
````

**Error intencionado: `TypeError`.** Los conjuntos no tienen índices. Por eso `land_cover_set[0]` no es válido.

````python
land_cover_set = {"forest", "water", "urban", "cropland"}
land_cover_set[0]
````

### Añadir elementos con `.add()`

`.add()` incorpora un elemento nuevo; si ya existe, no lo añade otra vez.

````python
land_cover_set = {"forest", "water", "urban", "cropland"}
land_cover_set.add("wetland")
print(sorted(land_cover_set))
````

### Comprobar la presencia de un elemento

El operador `in` indica si un valor pertenece al conjunto y devuelve `True` o `False`.

````python
land_cover_set = {"forest", "water", "urban", "cropland"}
print("Forest in set:", "forest" in land_cover_set)
print("Grassland in set:", "grassland" in land_cover_set)
````

### Comparar conjuntos

Se pueden comparar conjuntos con cuatro operaciones:

- **Intersección**: valores presentes en ambos conjuntos.
- **Unión**: valores presentes en al menos uno de los conjuntos, sin duplicados.
- **Diferencia**: valores del primer conjunto que no están en el segundo.
- **Diferencia simétrica**: valores que aparecen en uno u otro conjunto, pero no en ambos.

El orden en que se muestran los resultados no es significativo.

#### Intersección

`.intersection()` devuelve las coberturas que aparecen en las dos áreas.

````python
area_a = {"forest", "water", "urban", "cropland"}
area_b = {"forest", "grassland", "cropland", "wetland"}
print(sorted(area_a.intersection(area_b)))
````

#### Unión

`.union()` devuelve todas las coberturas diferentes presentes en cualquiera de las áreas.

````python
area_a = {"forest", "water", "urban", "cropland"}
area_b = {"forest", "grassland", "cropland", "wetland"}
print(sorted(area_a.union(area_b)))
````

#### Diferencia

`.difference()` devuelve las coberturas del área A que no aparecen en el área B. El orden de los conjuntos importa en esta operación.

````python
area_a = {"forest", "water", "urban", "cropland"}
area_b = {"forest", "grassland", "cropland", "wetland"}
print(sorted(area_a.difference(area_b)))
````

#### Diferencia simétrica

`.symmetric_difference()` devuelve las coberturas que aparecen solo en una de las dos áreas.

````python
area_a = {"forest", "water", "urban", "cropland"}
area_b = {"forest", "grassland", "cropland", "wetland"}
print(sorted(area_a.symmetric_difference(area_b)))
````

## Resumen

| Tipo | Sintaxis habitual | Conserva el orden | Admite duplicados | Permite modificar sus elementos |
|---|---|---|---|---|
| Lista (`list`) | `["a", "b"]` | Sí | Sí | Sí |
| Tupla (`tuple`) | `("a", "b")` | Sí | Sí | No directamente¹ |
| Conjunto (`set`) | `{"a", "b"}` | No tiene orden fijo | No | Se pueden añadir o eliminar valores² |

¹ Una tupla puede contener listas que sí se pueden modificar.  
² Un conjunto es mutable, pero no permite acceder a sus elementos por posición.

**Ideas clave:** los índices de las listas y las tuplas comienzan en `0`; `in` comprueba pertenencia; `len()` cuenta elementos; las cadenas de texto son inmutables; y los conjuntos permiten trabajar con valores únicos.
