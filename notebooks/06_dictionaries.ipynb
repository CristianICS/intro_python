# Diccionarios

Los diccionarios permiten almacenar información relacionada entre sí. Por ejemplo, el nombre de una capa junto con su tipo de geometría y su sistema de referencia.

Objetivos de aprendizaje:

- Aprender a usar diccionarios en el contexto de la información geográfica
- Conocer sus reglas y limitaciones

## ¿Qué son los diccionarios?

Un diccionario almacena información en pares *clave-valor*: cada **clave** se asocia a un **valor**, y la clave sirve para recuperar ese valor, igual que una palabra permite encontrar su definición en un diccionario de papel.

Una lista se consulta por **posición** (`lista[0]`); un diccionario se consulta por **clave** (`diccionario["clave"]`). En el trabajo con datos geográficos son omnipresentes: los atributos de una entidad, los metadatos de una capa o los códigos de los sistemas de referencia son pares clave-valor.

Reglas básicas:

- Las claves son **únicas**. Si se repite una clave, prevalece el último valor asignado.
- Las claves deben ser **inmutables**: cadenas de texto, números o tuplas. Una lista no puede ser clave.
- Los valores pueden ser de cualquier tipo (números, cadenas, listas, otros diccionarios…).
- Desde Python 3.7, el diccionario conserva el orden de inserción de los pares. Aun así, **no** se accede a los elementos por posición.

### Sintaxis general

Un diccionario se escribe entre llaves, con los pares `clave: valor` separados por comas:

```python
dictionary_name = {key_1: value_1, key_2: value_2, key_3: value_3}
```

Cuando las claves o los valores son largos, es habitual escribir un par por línea, lo que mejora la legibilidad:

```python
dictionary_name = {
    key_1: value_1,
    key_2: value_2,
    key_3: value_3,
}
```

### Ejemplo

Un diccionario que asocia el nombre de un sistema de referencia de coordenadas (CRS) con su código EPSG. El valor se obtiene escribiendo el nombre del diccionario y la clave entre corchetes:

````python
epsg_codes = {
    "WGS 84": 4326,
    "Web Mercator": 3857,
    "ETRS89 / UTM 30N": 25830,
}

print(epsg_codes["WGS 84"])
print(epsg_codes["ETRS89 / UTM 30N"])
````

Para mostrar todo el contenido no hace falta repetir una instrucción por cada par. Un bucle `for` sobre `.items()` recorre el diccionario completo:

````python
for crs_name, code in epsg_codes.items():
    print(f"{crs_name} -> EPSG:{code}")
````

Notas sobre este bucle:
- `.items()` devuelve los pares como tuplas `(clave, valor)`. En cada vuelta, Python *desempaqueta* la tupla en las dos variables del bucle: la primera recibe la clave y la segunda, el valor.
- Los nombres de esas variables son libres; conviene elegir nombres que describan su contenido (`crs_name`, `code`).
- `f"..."` es una *f-string*: lo que va entre llaves se sustituye por el valor de la variable.

Sintaxis general:

```python
for key_name, value_name in dictionary_name.items():
    print(key_name)    # la clave
    print(value_name)  # el valor asociado
```

### Claves inexistentes

Consultar una clave que no existe provoca un error `KeyError`. Hay dos formas de evitarlo:
- El operador `in` comprueba si una clave está en el diccionario (devuelve `True` o `False`).
- El método `.get()` devuelve `None` (o un valor por defecto indicado como segundo argumento) si la clave no existe.

````python
epsg_codes = {
    "WGS 84": 4326,
    "Web Mercator": 3857,
    "ETRS89 / UTM 30N": 25830,
}

# print(epsg_codes["ED50"])  # KeyError: 'ED50'

print("ED50" in epsg_codes)
print(epsg_codes.get("ED50"))
print(epsg_codes.get("ED50", "not found"))
````

## Operaciones habituales

Las operaciones más frecuentes son añadir pares, modificar valores o claves y eliminar pares.

### Añadir pares clave-valor

Se escribe el nombre del diccionario, la nueva clave entre corchetes y se le asigna el valor. Se puede partir de un diccionario vacío (`{}`) y rellenarlo par a par:

````python
layers = {}

layers["roads"] = "line"
layers["cities"] = "point"
layers["lakes"] = "polygon"

print(layers)
````

En el cuaderno sobre los conjuntos o `set`, se explicó que un conjunto vacío solo podía crearse con `set()`. El motivo, utilizar `{}` crearía un diccionario vacío.

### Crear un diccionario a partir de dos listas

Con `zip()` (visto en el cuaderno de bucles) se pueden emparejar dos listas paralelas, por ejemplo, una columna de nombres y otra de valores de una tabla de atributos. La función `dict()` convierte esos pares en un diccionario, igual que `list()` convierte un objeto en lista o `str()` en texto: la primera lista aporta las claves y la segunda, los valores.

````python
station_names = ["Madrid", "Granada", "Santander", "Madrid"]
altitudes = [667, 738, 15, 800]

station_altitudes = dict(zip(station_names, altitudes))
print(station_altitudes)
````

Conviene tener en cuenta dos detalles:

- Las dos listas deben estar en el mismo orden y tener la misma longitud. Si una es más larga, `zip()` ignora los elementos sobrantes sin avisar.
- Si hay claves repetidas, prevalece el último valor, como en cualquier diccionario (en el diccionario anterior, `Madrid` contiene el último valor de altura `800`).

### Modificar valores

Se usa la misma sintaxis que para añadir: si la clave ya existe, su valor se sustituye.

````python
layers["lakes"] = "multipolygon"
print(layers)
````

### Eliminar pares clave-valor

Hay dos opciones, y ambas eliminan la clave y su valor a la vez:

- `del diccionario[clave]` elimina el par.
- `diccionario.pop(clave)` elimina el par y además **devuelve** el valor eliminado, que puede guardarse en una variable.

````python
del layers["cities"]
print(layers)

removed_value = layers.pop("roads")
print(removed_value)
print(layers)
````

### Modificar claves

Una clave no puede editarse directamente, porque identifica a su valor. Para cambiarla, se crea la clave nueva con el valor de la antigua y se elimina esta. `pop()` permite hacerlo en una sola línea, ya que elimina la clave antigua y devuelve su valor:

````python
# Tipo en la clave: "raods"
layers = {"raods": "line"}

layers["roads"] = layers.pop("raods")

print(layers)
````

## Recorrer un diccionario

Un bucle puede recorrer un diccionario de tres formas:

- por **pares clave-valor** (`.items()`);
- por **claves** (`.keys()`, o el propio diccionario);
- por **valores** (`.values()`).

Los ejemplos de esta sección usan un diccionario con la altitud (en metros, valores aproximados) de varios picos de la península ibérica.

### Pares clave-valor

Ya visto con `.items()`. Este método devuelve una vista de tuplas `(clave, valor)` que puede convertirse en lista con `list()` para inspeccionarla:

````python
elevations = {
    "Mulhacén": 3479,
    "Aneto": 3404,
    "Veleta": 3396,
    "Almanzor": 2592,
}

print(list(elevations.items()))

for peak, height in elevations.items():
    print(f"{peak}: {height} m")
````

### Claves

Recorrer solo las claves es el comportamiento por defecto de un bucle sobre un diccionario, por lo que `.keys()` es opcional (ambas versiones son equivalentes):

````python
for peak in elevations:
    print(peak)

for peak in elevations.keys():
    print(peak)
````

### Valores

Cuando solo importan los valores, `.values()` permite recorrerlos o aplicarles funciones como `max()`, `min()` o `sum()`:

````python
for height in elevations.values():
    print(height)

print("Highest:", max(elevations.values()))
print("Mean:", sum(elevations.values()) / len(elevations))
````

Se puede saber cuál es la clave que corresponde a un valor aplicando un filtro con un bucle *for*. Podría haber dos claves con el mismo valor de altura, por lo que la mejor opción es guardar las claves seleccionadas en una lista.

Esto sucede porque los diccionarios están optimizados para buscar valores por su clave, y no al revés.

````python
elevations = {"Everest": 8849, "K2": 8611, "Kangchenjunga": 8586}
max_height = max(elevations.values())

highest_peaks = []
for peak, height in elevations.items():
    if height == max_height:
        highest_peaks.append(peak)

print(highest_peaks)
````

La operación anterior se puede realizar con una comprensión. Comprobar que la función `max()` permite incluir un segundo argumento. Este, al igual que en el método `get()`, incluye el valor por defecto cuando el diccionario está vacío. En este ejemplo no sucede, pero en ocasiones un programa podría leer un diccionario vacío.

````python
elevations = {"Everest": 8849, "K2": 8611, "Kangchenjunga": 8586}
max_height = max(elevations.values(), default=None)

highest_peaks = [
    peak
    for peak, height in elevations.items()
    if height == max_height
]
print(highest_peaks)
````

### Comprensiones de diccionarios

Igual que una comprensión de listas (cuaderno `05_loops.ipynb`) crea una lista en una sola expresión, una **comprensión de diccionario** crea un diccionario. Se escribe entre llaves y cada elemento es un par `clave: valor`:

```python
{key: value for element in collection}
```

El siguiente ejemplo convierte las altitudes de metros a pies. La segunda comprensión añade un `if` para conservar solo los picos de más de 8600 m:

````python
elevations = {"Everest": 8849, "K2": 8611, "Kangchenjunga": 8586}

elevations_ft = {
    peak: round(height * 3.28084)
    for peak, height in elevations.items()
}
print(elevations_ft)

highest = {
    peak: height
    for peak, height in elevations.items()
    if height > 8600
}
print(highest)
````

Una comprensión también puede recorrer dos listas con `zip()`. El resultado equivale a `dict(zip(...))`, pero permite transformar o filtrar los valores al crear el diccionario:

````python
station_names = ["Madrid", "Granada", "Santander"]
altitudes = [667, 738, 15]

station_altitudes_km = {
    name: altitude / 1000
    for name, altitude in zip(station_names, altitudes)
}
print(station_altitudes_km)
````

## Orden

Desde Python 3.7, los pares se recorren en el orden en que se insertaron. Si se necesita otro orden, la función `sorted()` devuelve una lista ordenada. Aplicada a las claves, las ordena alfabéticamente:

````python
elevations = {"Everest": 8849, "K2": 8611, "Kangchenjunga": 8586}

for peak in sorted(elevations):
    print(f"{peak}: {elevations[peak]} m")

# Valores de mayor a menor
print(sorted(elevations.values(), reverse=True))
````

`sorted()` no modifica el diccionario: solo cambia el orden del recorrido en ese bucle.

## Anidamiento

El *anidamiento* consiste en colocar una lista o un diccionario dentro de otro diccionario (o lista). Permite modelar información más compleja, como la que manejan habitualmente los SIG: una entidad con varios atributos, una capa con sus metadatos o una serie de mediciones por estación.

### Listas dentro de un diccionario

Un valor puede ser una lista. En este ejemplo, cada estación meteorológica (clave) tiene una lista de temperaturas diarias en °C (valor):

````python
temperatures = {
    "Madrid": [12.1, 14.3, 13.8, 11.5],
    "Granada": [15.0, 16.2, 14.9],
    "Santander": [10.4, 9.8, 11.2, 10.9, 10.1],
}

# Acceso directo a una lista
print(temperatures["Granada"])

# Primer valor de esa lista (dos accesos encadenados)
print(temperatures["Granada"][0])
````

Para procesar todas las estaciones se combinan dos conceptos ya conocidos: un bucle sobre el diccionario y, dentro de él, trabajo con la lista de cada estación (en este caso, con `sum()` y `len()`):

````python
for station, readings in temperatures.items():
    mean_temperature = sum(readings) / len(readings)

    message = f"{len(readings)} readings, mean {round(mean_temperature, 1)} °C"
    print(f"{station}: {message}")
````

Si hay que tratar cada valor de la lista, se añade un segundo bucle.

Nota: la instrucción `break` hace que termine el bucle `for` cuando llega a ella. En el siguiente ejemplo se utiliza para evitar mostrar en pantalla los valores de todas las estaciones.

````python
# Si hay que tratar cada valor de la lista, se añade un segundo bucle
for station, readings in temperatures.items():
    print(f"\nReadings for {station}:")
    for reading in readings:
        print(reading)

    break
````

### Diccionarios dentro de un diccionario

Es el tipo de anidamiento más potente. Aquí, cada clave es el nombre de una capa y cada valor es otro diccionario con sus metadatos:

````python
layers = {
    "cities": {"geometry": "point", "epsg": 4326, "features": 120},
    "roads": {"geometry": "line", "epsg": 25830, "features": 3400},
    "lakes": {"geometry": "polygon", "epsg": 4326, "features": 58},
}

# Cada corchete "abre" un nivel: primero la capa y luego el metadato
print(layers["roads"]["epsg"])

# Modificar un metadato
layers["cities"]["features"] = 125
print(layers["cities"])
````

Para recorrerlo se usan dos bucles anidados:
- El bucle externo recorre las capas (claves del diccionario principal).
- Cada valor `info` es un diccionario, de modo que el bucle interno recorre sus pares clave-valor.

Así, si se añaden más metadatos, el código no necesita cambios.

````python
for layer_name, info in layers.items():
    print(f"\nLayer: {layer_name}")
    for key, value in info.items():
        print(f"- {key}: {value}")
````

Si se necesita un formato distinto según la clave, se puede añadir un `if` dentro del bucle interno:

````python
for layer_name, info in layers.items():
    print(f"\nLayer: {layer_name.title()}")
    for key, value in info.items():
        if key == "epsg":
            print(f"- CRS: EPSG:{value}")
        else:
            print(f"- {key}: {value}")
    break
````

### Una nota sobre el anidamiento

Los datos geoespaciales suelen estar anidados a varios niveles. Un ejemplo es el formato GeoJSON, que Python lee como diccionarios y listas combinados:

````python
feature = {
    "type": "Feature",
    "properties": {"name": "Madrid", "population": 3300000},
    "geometry": {"type": "Point", "coordinates": [-3.7038, 40.4168]},
}

longitude = feature["geometry"]["coordinates"][0]
latitude = feature["geometry"]["coordinates"][1]
print(longitude, latitude)
````

Conviene recordar dos ideas:
- GeoJSON (y el estándar en general) ordena las coordenadas como **(longitud, latitud)**, es decir, (x, y). Muchas herramientas y personas escriben primero la latitud, por lo que este orden es una fuente habitual de errores.
- Con más de dos niveles, los accesos encadenados se vuelven difíciles de leer. Una buena práctica es guardar los niveles intermedios en variables (`geometry = feature["geometry"]`). Para volúmenes grandes de datos existen herramientas específicas (bases de datos espaciales, librerías como *GeoPandas*) que se verán más adelante.

La estructura de un GeoJSON incluye cada elemento `feature` en el siguiente diccionario.

````python
{
  "type": "FeatureCollection",
  "features": [feature]
}
````

## Resumen

**Qué es un diccionario.** Una colección de pares **clave-valor** escrita entre llaves. Se consulta por **clave** (`d["clave"]`), no por posición como las listas.

Reglas básicas:

- Las claves son **únicas** (si se repite, prevalece el último valor) e **inmutables** (cadenas, números o tuplas; nunca listas).
- Los valores pueden ser de **cualquier tipo**, incluidas listas y otros diccionarios.
- Desde Python 3.7 conservan el **orden de inserción**.
- `{}` crea un diccionario vacío; para un conjunto vacío hay que usar `set()`.

Operaciones habituales:

| Operación | Sintaxis |
| --- | --- |
| Añadir o modificar un valor | `d[clave] = valor` |
| Eliminar un par | `del d[clave]` |
| Eliminar y recuperar el valor | `valor = d.pop(clave)` |
| Renombrar una clave | `d[nueva] = d.pop(antigua)` |
| Comprobar si existe una clave | `clave in d` |
| Consultar sin riesgo de error | `d.get(clave, valor_por_defecto)` |
| Crear diccionario a partir de dos listas | `dict(zip(claves, valores))` |
| Crear un diccionario transformando o filtrando otro | `{k: v for k, v in d.items() if condición}` |

Recorrer un diccionario:

- `for k, v in d.items()`: pares clave-valor (la opción más habitual).
- `for k in d` (o `d.keys()`): solo claves.
- `for v in d.values()`: solo valores, útil con `max()`, `min()` o `sum()`.
- `sorted(d)` recorre las claves en orden alfabético sin modificar el diccionario.

**Buscar por valor.** Los diccionarios están pensados para buscar valores a partir de claves, no al revés. Para encontrar las claves de un valor se filtra con un bucle y se guardan en una **lista**, porque varias claves pueden compartir valor.

Anidamiento:

- Con listas o diccionarios como valores se modelan datos complejos (atributos, metadatos, series de mediciones).
- Se accede encadenando corchetes: `d["roads"]["epsg"]`.
- Se recorren con **bucles anidados**: el externo para las claves principales y el interno para el contenido de cada valor.
- Con más de dos niveles, conviene guardar los niveles intermedios en variables para mantener el código legible.
