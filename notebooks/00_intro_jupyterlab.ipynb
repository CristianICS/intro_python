## Crear celdas

Una de las tareas que más se repite cuando se trabaja con cuadernos es la creación de celdas, tanto de código como de texto. La celda actual solo incluye texto plano, con formato Markdown.

Existen tres opciones principales para crear una celda de código:

- Pulsar sobre el símbolo `+` en el menú de opciones del cuaderno actual ![](../images/exercises/jupyter_notebook_menu_icon.png)
- Seleccionar el icono de crear una celda (arriba o abajo) en la parte superior derecha de una celda activa: ![](../images/exercises/create_cell_jupyter_icon_inside.png)
- Pulsando la tecla `A` (crea la celda encima de la casilla seleccionada) o `B` (la crea debajo)

Las celdas se crean debajo de la celda sobre la que se encuentra el "foco", señalada por la barra lateral azul.

![Vista de una celda con foco.](../images/exercises/focused_cell.png)

## Ejecutar las celdas

Una vez que se ha creado una celda y añadido contenido, se pueden ejecutar mediante:

- `Ctrl + Enter`
- Icono *Play* en el menú de opciones del cuaderno
- `Shift + Enter`

Las dos últimas opciones ejecutan la celda y **se desplazan a la siguiente** en el orden del cuaderno, de arriba hacia abajo.

Las celdas de texto, cuando se ejecutan, muestran el texto con las etiquetas de Markdown aplicadas

Existe además una opción para ejecutar el documento completo en el menú de opciones del cuaderno actual ![](../../images/exercises/jupyter_notebook_menu_icon.png), los dos triángulos apuntando hacia la derecha (símbolo de avance rápido). Esta opción es la más parecida a ejecutar un código de Python desde un archivo `.py`, evitando comportamientos anómalos por no haber ejecutado una celda en el orden correcto.

> Nota: La opción anterior reiniciará el *kernel*, lo que significa que el resultado del código que se haya ejecutado y ya no esté en el cuaderno se perderá. Ver la [sección Kernel](#Kernel).

## Modos disponibles

A la hora de modificar una celda, tiene que estar en modo edición.

Entrar en modo de edición:

- pulsar `Enter`
- hacer clic sobre la celda una vez (si es de código) o dos veces (de texto)

Salir del modo edición:

- pulsar la tecla `Esc`
- en algunas configuraciones, `Ctrl + M` o `Ctrl + [`
- Hacer clic fuera de la celda (en el margen o en alguna otra zona neutra, sin celdas, del cuaderno)

## Modificar celdas

**Fuera del modo de edición** (ver la sección anterior) se pueden utilizar una serie de comandos para modificar las celdas.

Se pueden desplazar arriba y abajo mediante los comandos

- `Ctrl + Shift + Up` (arriba)
- `Ctrl + Shift + Down` (abajo)

Otra forma es pulsar sobre las flechas que aparecen en la parte superior izquierda de una celda al entrar en el modo de edición.

> Nota: Se pueden mover varias celdas seleccionándolas con `Shift + Up` o `Shift + Down`. Una vez seleccionadas, moverlas con los comandos anteriores.

Además de mover celdas, se pueden eliminar pulsando `D, D` y cortarlas con la tecla `X`. La última opción corta y elimina las celdas, por lo que pueden pegarse pulsando `V`. Por defecto, se pegan debajo de la celda en la que se encuentra el foco (ver la [Sección Crear celdas](#Crear-celdas)).

Las celdas pueden copiarse pulsando `C` fuera del modo de edición.



## Hacer y deshacer cambios

Si se ha eliminado una celda por accidente, se puede deshacer el cambio pulsando `Z`. Al contrario, se puede rehacer la acción deshecha pulsando `Shift + Z`.

Estas opciones también están disponibles dentro del menú superior *Edit*.

Si se está **editando una celda** (modo edición), el comando deshacer es `Ctrl + Z` y el comando rehacer `Ctrl + Y`.

## Opciones generales

Para acceder al conjunto de opciones de configuración, abrir el panel de opciones con `Ctrl + ,`. También puede desplegarse desde la opción del menú superior *Settings* > *Settings editor*.

Dentro de este panel, las opciones para configurar las celdas de los cuadernos se encuentran en la sección *Notebook* (panel izquierdo). Algunas de las más utilizadas son:

- *Default cell type* define el tipo de celda por defecto.
- Dentro de la sección *Code Cell Configuration*, que marca el comportamiento de las celdas de código:
    - Marcar *Auto closing brackets*. Permite cerrar los paréntesis y corchetes al escribir código de Python.
    - *Indentation unit* marca los espacios utilizados al utilizar la sangría (se recomienda utilizar 4 para código de Python)

## Kernel

Cuando un cuaderno es abierto por JupyterLab, la parte visible se convierte en una página web (código HTML), pero por debajo una sesión se activa para poder ejecutar el código. Cada una de estas sesiones se conoce con el nombre de *kernels*. La conexión entre la página web y el *kernel* se realiza por el servidor, ejecutado en la terminal dentro del entorno de conda.

## Atajos dentro de las celdas

Se han visto comandos que permiten modificar el contenido del cuaderno en el modo de vista (es decir, sin ninguna celda en modo de edición). Existen atajos de teclado que pueden utilizarse cuando una celda está en modo de edición.

- `Alt + Up` o `Alt + Down` mueven una línea arriba o abajo
- `Shift + Alt + Up` o `Shift + Alt + Down` duplican la línea y la mueven arriba o abajo
- `Ctrl + D` elimina la línea completa
- `Ctrl + Ç` (`Ctrl + /` en teclados ingleses) comenta la línea (no se mostrará si es una celda de texto, no se ejecutará si es una celda de código)
- `Tab` y `Shift + Tab` aplican o quitan la sangría en una línea

## Terminar la sesión

La sesión actual se refiere al servidor iniciado en el entorno de conda y que se mantiene activo en todo momento. Si se cierra la pestaña del navegador donde se está ejecutando, el servidor no se cierra.

Existen dos opciones para cerrar o terminar el servidor:

Pulsar `Ctrl + C` en la terminal que ejecuta el entorno de conda. Se verá que vuelve a aparecer al inicio de la línea el nombre del entorno y la ruta donde se ejecuta la consola, preparada para volver a incluir una nueva línea:

```text
(hello_world) PS D:/intro_python>
```

Una segunda forma de terminar con la sesión es pulsando sobre *Shut down*, desde *File* en el menú superior. Esta opción termina el servidor, y aparece un mensaje de aviso en JupyterLab que avisa de que el servidor se ha cerrado. A veces la aplicación JupyterLab se cierra automáticamente y otras no.
